```python
import pandas as pd
import os

# பாதை அமைப்பு (Path setup)
base_path = '/kaggle/input/datasets/sudalairajkumar/tamil-nlp/'

# நாம் பயன்படுத்த விரும்பும் Train கோப்புகள்
files_to_load = [
    'tamil_news_train.csv',
    'tamil_thirukkural_train.csv',
    'tamil_movie_reviews_train.csv'
]

# அனைத்து தமிழ் உரைகளையும் ஒரே பட்டியலில் சேகரிப்போம்
all_tamil_text = []

print("கோப்புகளை ஏற்றுகிறோம்...")

for file_name in files_to_load:
    full_path = os.path.join(base_path, file_name)

    # கோப்பு இருக்கிறதா என்று சரிபார்க்கவும்
    if os.path.exists(full_path):
        try:
            # கோப்பை ஏற்றவும்
            df = pd.read_csv(full_path)
            print(f"ஏற்றப்பட்டது: {file_name} - அளவு: {df.shape}")

            # கோப்பு வகையைப் பொறுத்து தமிழ் நெடுவரிசையை (Column) தேர்வு செய்யவும்
            # செய்தி கோப்புகளுக்கு 'NewsInTamil', திருக்குறளுக்கு 'Verse' அல்லது 'explanation' இருக்கலாம்.
            # நெடுவரிசை பெயர்களைப் பார்த்து தானாக தேர்வு செய்யும் முறை:

            text_col = None
            if 'NewsInTamil' in df.columns:
                text_col = 'NewsInTamil'
            elif 'Review' in df.columns: # Movie reviews
                text_col = 'Review'
            elif 'Verse' in df.columns: # Thirukkural
                text_col = 'Verse'

            if text_col:
                # NaN மதிப்புகளை நீக்கி, உரையை பட்டியலில் சேர்க்கவும்
                texts = df[text_col].dropna().tolist()
                all_tamil_text.extend(texts)
                print(f"   -> '{text_col}' நெடுவரிசையிலிருந்து {len(texts)} வரிகள் சேர்க்கப்பட்டன.")
            else:
                print(f"   -> கவனம்: {file_name}-ல் தமிழ் உரை நெடுவரிசை கண்டறியப்படவில்லை. நெடுவரிசைகள்: {df.columns.tolist()}")

        except Exception as e:
            print(f"பிழை {file_name}: {e}")
    else:
        print(f"கோப்பு கிடைக்கவில்லை: {file_name}")

print(f"\nமொத்தம் சேகரிக்கப்பட்ட தமிழ் வரிகள்: {len(all_tamil_text)}")

# முதல் 3 உதாரணங்களைக் காட்டவும்
print("\nமுதல் 3 உதாரணங்கள்:")
for i, text in enumerate(all_tamil_text[:3]):
    print(f"{i+1}. {text[:100]}...") # முதல் 100 எழுத்துக்கள்
```

```python
import pandas as pd
import os
import re

# பாதை அமைப்பு
base_path = '/kaggle/input/datasets/sudalairajkumar/tamil-nlp/'

# தற்காலிக பட்டியல்
all_tamil_text = []

# ---------------------------------------------------------
# 1. Tamil News Data
# ---------------------------------------------------------
file_path = os.path.join(base_path, 'tamil_news_train.csv')
if os.path.exists(file_path):
    df = pd.read_csv(file_path)
    # 'NewsInTamil' நெடுவரிசையை சேர்க்கவும்
    all_tamil_text.extend(df['NewsInTamil'].dropna().tolist())
    print(f"News: {len(df)} வரிகள் சேர்க்கப்பட்டன.")

# ---------------------------------------------------------
# 2. Tamil Movie Reviews Data
# ---------------------------------------------------------
file_path = os.path.join(base_path, 'tamil_movie_reviews_train.csv')
if os.path.exists(file_path):
    df = pd.read_csv(file_path)
    # இங்கே 'ReviewInTamil' என்ற பெயரில் உள்ளது
    all_tamil_text.extend(df['ReviewInTamil'].dropna().tolist())
    print(f"Movie Reviews: {len(df)} வரிகள் சேர்க்கப்பட்டன.")

# ---------------------------------------------------------
# 3. Thirukkural Data (திருக்குறள்)
# ---------------------------------------------------------
file_path = os.path.join(base_path, 'tamil_thirukkural_train.csv')
if os.path.exists(file_path):
    df = pd.read_csv(file_path)
    # திருக்குறளில் 'kural' (பாடல்) மற்றும் 'explanation' (விளக்கம்) இரண்டும் முக்கியம்
    kurals = df['kural'].dropna().tolist()
    explanations = df['explanation'].dropna().tolist()

    all_tamil_text.extend(kurals)
    all_tamil_text.extend(explanations)
    print(f"Thirukkural: {len(kurals) + len(explanations)} வரிகள் சேர்க்கப்பட்டன.")

# ---------------------------------------------------------
# தரவு சுத்தம் செய்தல் (Data Cleaning)
# ---------------------------------------------------------
print("\nதரவு சுத்தம் செய்கிறோம்...")

def clean_tamil_text(text):
    if not isinstance(text, str):
        return ""

    # 1. URL-களை அகற்றுதல்
    text = re.sub(r'http\S+', '', text)

    # 2. வெறும் ஆங்கில எழுத்துக்கள் மற்றும் எண்கள் மட்டும் இருந்தால் அகற்றலாம் (விருப்பம்)
    # ஆனால் LLM-க்கு சில ஆங்கில வார்த்தைகள் (உதா: 'T20', 'GDP') தேவைப்படலாம்.
    # எனவே நாம் வெறும் இடைவெளிகளை (whitespaces) மட்டும் சரி செய்வோம்.

    # 3. அதிக இடைவெளிகளை அகற்றுதல்
    text = re.sub(r'\s+', ' ', text).strip()

    return text

# அனைத்து உரைகளையும் சுத்தம் செய்தல்
cleaned_text_list = [clean_tamil_text(t) for t in all_tamil_text if clean_tamil_text(t)]

# காலியான வரிகளை நீக்குதல்
cleaned_text_list = [t for t in cleaned_text_list if len(t) > 10] # 10 எழுத்துகளுக்கு மேல் உள்ளவை மட்டும்

print(f"சுத்தம் செய்யப்பட்ட மொத்த வரிகள்: {len(cleaned_text_list)}")

# ---------------------------------------------------------
# கோப்பாக சேமித்தல் (Save to File)
# ---------------------------------------------------------
output_file = '/kaggle/working/tamil_corpus.txt'

with open(output_file, 'w', encoding='utf-8') as f:
    for line in cleaned_text_list:
        f.write(line + '\n')

print(f"\nவெற்றி! அனைத்து தரவுகளும் சேமிக்கப்பட்டன: {output_file}")

# கோப்பின் அளவை சரிபார்க்கவும்
file_size = os.path.getsize(output_file)
print(f"கோப்பின் அளவு: {file_size / 1024:.2f} KB")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import sentencepiece as spm
import os

# =========================================================
# 0. SAMPLE DATA CREATION (புதிய தரவு உருவாக்குதல்)
# =========================================================
print("📝 Sample Tamil Corpus உருவாக்குகிறேன்...")

tamil_samples = [
    "தமிழ் மொழி மிகவும் பழமையான மொழி.",
    "இது இந்தியாவில் பேசப்படும் முக்கிய மொழிகளில் ஒன்று.",
    "தமிழ் இலக்கியம் பல்லாயிரம் ஆண்டுகள் பழமையானது.",
    "சங்க கால இலக்கியங்கள் தமிழின் சிறப்பை காட்டுகின்றன.",
    "தமிழ் மக்கள் தங்கள் மொழியை பெருமையாக கருதுகின்றனர்.",
    "உலகின் பல பகுதிகளிலும் தமிழ் பேசப்படுகிறது.",
    "தமிழ் மொழி இனிமையானது.",
    "இன்று காலை நல்ல வானிலை.",
    "இந்தியா ஒரு அழகான நாடு.",
    "காலை வணக்கம் நண்பர்களே.",
    "தமிழ் நாடு இந்தியாவின் ஒரு மாநிலம்.",
    "சென்னை தமிழ் நாட்டின் தலைநகர்.",
    "தமிழ் மொழி இசையானது.",
    "இலக்கணம் தமிழின் அடிப்படை.",
    "தமிழ் பேசுவோம் தமிழ் வளர்ப்போம்.",
    "நல்ல காலை வணக்கம்.",
    "இனிய பகல் வணக்கம்.",
    "இனிய இரவு வணக்கம்.",
    "தமிழ் மொழி வளர்க.",
    "அறம் செய விரும்பு."
]

# Data augmentation - ஒவ்வொரு வரியை 50 மடங்கு மீண்டும் எழுதுதல்
corpus_text = "\n".join(tamil_samples * 50)

corpus_path = '/kaggle/working/tamil_corpus.txt'
with open(corpus_path, 'w', encoding='utf-8') as f:
    f.write(corpus_text)

print(f"✅ Corpus உருவாக்கப்பட்டது!")
print(f"📏 அளவு: {len(corpus_text)} characters")

# Tokenizer உருவாக்குதல்
print("\n🔤 Tokenizer உருவாக்குகிறேன்...")
tokenizer_prefix = '/kaggle/working/tamil_tokenizer'

spm.SentencePieceTrainer.train(
    input=corpus_path,
    model_prefix=tokenizer_prefix,
    vocab_size=300,  # சிறிய corpus-க்கு சிறிய vocab
    character_coverage=1.0,
    model_type='bpe',
    num_threads=1
)

tokenizer_path = '/kaggle/working/tamil_tokenizer.model'
print(f"✅ Tokenizer உருவாக்கப்பட்டது!")

# =========================================================
# 1. CONFIGURATION
# =========================================================
config = {
    "batch_size": 4,       # சிறிய corpus-க்கு சிறிய batch
    "block_size": 32,    # சிறிய context
    "n_embd": 64,        # சிறிய embedding
    "n_head": 2,
    "n_layer": 2,
    "dropout": 0.1,
    "learning_rate": 3e-4,
    "epochs": 50         # அதிக epochs (சிறிய data)
}

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"\n💻 சாதனம்: {device}")

# =========================================================
# 2. DATASET
# =========================================================
class TamilDataset(Dataset):
    def __init__(self, file_path, tokenizer_path, block_size):
        self.sp = spm.SentencePieceProcessor()
        self.sp.load(tokenizer_path)

        with open(file_path, 'r', encoding='utf-8') as f:
            text = f.read()

        self.tokens = self.sp.encode(text, out_type=int)
        self.block_size = block_size
        self.data_len = max(1, len(self.tokens) // block_size - 1)

        print(f"📊 Tokens: {len(self.tokens)}, Training samples: {self.data_len}")

    def __len__(self):
        return self.data_len

    def __getitem__(self, idx):
        start = idx * self.block_size
        end = start + self.block_size
        x = torch.tensor(self.tokens[start:end], dtype=torch.long)
        y = torch.tensor(self.tokens[start+1:end+1], dtype=torch.long)
        return x, y

dataset = TamilDataset(corpus_path, tokenizer_path, config['block_size'])
train_loader = DataLoader(dataset, batch_size=config['batch_size'], shuffle=True)

# =========================================================
# 3. MODEL ARCHITECTURE
# =========================================================
class Head(nn.Module):
    def __init__(self, head_size):
        super().__init__()
        self.key = nn.Linear(config['n_embd'], head_size, bias=False)
        self.query = nn.Linear(config['n_embd'], head_size, bias=False)
        self.value = nn.Linear(config['n_embd'], head_size, bias=False)
        self.register_buffer('tril', torch.tril(torch.ones(config['block_size'], config['block_size'])))
        self.dropout = nn.Dropout(config['dropout'])

    def forward(self, x):
        B,T,C = x.shape
        k = self.key(x)
        q = self.query(x)
        wei = q @ k.transpose(-2,-1) * C**-0.5
        wei = wei.masked_fill(self.tril[:T, :T] == 0, float('-inf'))
        wei = F.softmax(wei, dim=-1)
        wei = self.dropout(wei)
        v = self.value(x)
        out = wei @ v
        return out

class MultiHeadAttention(nn.Module):
    def __init__(self, num_heads, head_size):
        super().__init__()
        self.heads = nn.ModuleList([Head(head_size) for _ in range(num_heads)])
        self.proj = nn.Linear(config['n_embd'], config['n_embd'])
        self.dropout = nn.Dropout(config['dropout'])

    def forward(self, x):
        out = torch.cat([h(x) for h in self.heads], dim=-1)
        out = self.proj(out)
        return out

class FeedFoward(nn.Module):
    def __init__(self, n_embd):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(n_embd, 4 * n_embd),
            nn.ReLU(),
            nn.Linear(4 * n_embd, n_embd),
            nn.Dropout(config['dropout']),
        )
    def forward(self, x):
        return self.net(x)

class Block(nn.Module):
    def __init__(self, n_embd, n_head):
        super().__init__()
        head_size = n_embd // n_head
        self.sa = MultiHeadAttention(n_head, head_size)
        self.ffwd = FeedFoward(n_embd)
        self.ln1 = nn.LayerNorm(n_embd)
        self.ln2 = nn.LayerNorm(n_embd)

    def forward(self, x):
        x = x + self.sa(self.ln1(x))
        x = x + self.ffwd(self.ln2(x))
        return x

class TamilLLM(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.token_embedding_table = nn.Embedding(vocab_size, config['n_embd'])
        self.position_embedding_table = nn.Embedding(config['block_size'], config['n_embd'])
        self.blocks = nn.Sequential(*[Block(config['n_embd'], config['n_head']) for _ in range(config['n_layer'])])
        self.ln_f = nn.LayerNorm(config['n_embd'])
        self.lm_head = nn.Linear(config['n_embd'], vocab_size)

    def forward(self, idx, targets=None):
        B, T = idx.shape
        tok_emb = self.token_embedding_table(idx)
        pos_emb = self.position_embedding_table(torch.arange(T, device=device))
        x = tok_emb + pos_emb
        x = self.blocks(x)
        x = self.ln_f(x)
        logits = self.lm_head(x)

        if targets is None:
            loss = None
        else:
            B, T, C = logits.shape
            logits = logits.view(B*T, C)
            targets = targets.view(B*T)
            loss = F.cross_entropy(logits, targets)
        return logits, loss

# =========================================================
# 4. TRAINING
# =========================================================
vocab_size = dataset.sp.get_piece_size()
model = TamilLLM(vocab_size).to(device)
optimizer = torch.optim.AdamW(model.parameters(), lr=config['learning_rate'])

print(f"\n🔧 Parameters: {sum(p.numel() for p in model.parameters())/1e6:.2f}M")
print(f"📅 Epochs: {config['epochs']}\n")

for epoch in range(config['epochs']):
    total_loss = 0
    model.train()

    for batch_idx, (x, y) in enumerate(train_loader):
        x, y = x.to(device), y.to(device)

        logits, loss = model(x, y)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

        if batch_idx % 10 == 0:
            print(f"Epoch {epoch+1}/{config['epochs']} | Batch {batch_idx}/{len(train_loader)} | Loss: {loss.item():.4f}")

    avg_loss = total_loss / len(train_loader)
    print(f"---- Epoch {epoch+1} Done. Avg Loss: {avg_loss:.4f} ----\n")

# Save
torch.save(model.state_dict(), '/kaggle/working/tamil_llm_model.pth')
print(f"✅ Model saved! Final Loss: {avg_loss:.4f}")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import sentencepiece as spm

# ---------------------------------------------------------
# 1. CONFIG (அதே அமைப்புகள்)
# ---------------------------------------------------------
config = {
    "block_size": 32,
    "n_embd": 64,
    "n_head": 2,
    "n_layer": 2,
    "dropout": 0.1
}
device = torch.device('cpu')

# ---------------------------------------------------------
# 2. MODEL ARCHITECTURE (மீண்டும் வரையறை)
# ---------------------------------------------------------
class Head(nn.Module):
    def __init__(self, head_size):
        super().__init__()
        self.key = nn.Linear(config['n_embd'], head_size, bias=False)
        self.query = nn.Linear(config['n_embd'], head_size, bias=False)
        self.value = nn.Linear(config['n_embd'], head_size, bias=False)
        self.register_buffer('tril', torch.tril(torch.ones(config['block_size'], config['block_size'])))
        self.dropout = nn.Dropout(config['dropout'])
    def forward(self, x):
        B,T,C = x.shape
        k = self.key(x)
        q = self.query(x)
        wei = q @ k.transpose(-2,-1) * C**-0.5
        wei = wei.masked_fill(self.tril[:T, :T] == 0, float('-inf'))
        wei = F.softmax(wei, dim=-1)
        wei = self.dropout(wei)
        v = self.value(x)
        out = wei @ v
        return out

class MultiHeadAttention(nn.Module):
    def __init__(self, num_heads, head_size):
        super().__init__()
        self.heads = nn.ModuleList([Head(head_size) for _ in range(num_heads)])
        self.proj = nn.Linear(config['n_embd'], config['n_embd'])
        self.dropout = nn.Dropout(config['dropout'])
    def forward(self, x):
        out = torch.cat([h(x) for h in self.heads], dim=-1)
        out = self.proj(out)
        return out

class FeedFoward(nn.Module):
    def __init__(self, n_embd):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(n_embd, 4 * n_embd), nn.ReLU(), nn.Linear(4 * n_embd, n_embd), nn.Dropout(config['dropout']))
    def forward(self, x): return self.net(x)

class Block(nn.Module):
    def __init__(self, n_embd, n_head):
        super().__init__()
        head_size = n_embd // n_head
        self.sa = MultiHeadAttention(n_head, head_size)
        self.ffwd = FeedFoward(n_embd)
        self.ln1 = nn.LayerNorm(n_embd)
        self.ln2 = nn.LayerNorm(n_embd)
    def forward(self, x):
        x = x + self.sa(self.ln1(x))
        x = x + self.ffwd(self.ln2(x))
        return x

class TamilLLM(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.token_embedding_table = nn.Embedding(vocab_size, config['n_embd'])
        self.position_embedding_table = nn.Embedding(config['block_size'], config['n_embd'])
        self.blocks = nn.Sequential(*[Block(config['n_embd'], config['n_head']) for _ in range(config['n_layer'])])
        self.ln_f = nn.LayerNorm(config['n_embd'])
        self.lm_head = nn.Linear(config['n_embd'], vocab_size)
    def forward(self, idx, targets=None):
        B, T = idx.shape
        tok_emb = self.token_embedding_table(idx)
        pos_emb = self.position_embedding_table(torch.arange(T, device=idx.device))
        x = tok_emb + pos_emb
        x = self.blocks(x)
        x = self.ln_f(x)
        logits = self.lm_head(x)
        if targets is None:
            loss = None
        else:
            B, T, C = logits.shape
            logits = logits.view(B*T, C)
            targets = targets.view(B*T)
            loss = F.cross_entropy(logits, targets)
        return logits, loss

# ---------------------------------------------------------
# 3. LOAD MODEL & TOKENIZER
# ---------------------------------------------------------
sp = spm.SentencePieceProcessor()
sp.load('/kaggle/working/tamil_tokenizer.model')

vocab_size = sp.get_piece_size()
model = TamilLLM(vocab_size)
model.load_state_dict(torch.load('/kaggle/working/tamil_llm_model.pth', map_location=device))
model = model.to(device)
model.eval()

print(f"✅ மாடல் ஏற்றப்பட்டது!")
print(f"📊 Vocab size: {vocab_size}")
print(f"🔧 Parameters: {sum(p.numel() for p in model.parameters())/1e6:.2f}M")

# ---------------------------------------------------------
# 4. TEXT GENERATION FUNCTION
# ---------------------------------------------------------
def generate_text(prompt, max_new_tokens=50, temperature=0.8):
    """உரை உருவாக்கும் செயல்பாடு"""
    tokens = sp.encode(prompt, out_type=int)
    idx = torch.tensor([tokens], dtype=torch.long).to(device)

    print(f"\n📝 Prompt: {prompt}")
    print("-" * 40)

    for _ in range(max_new_tokens):
        idx_cond = idx[:, -config['block_size']:]

        with torch.no_grad():
            logits, _ = model(idx_cond)

        logits = logits[:, -1, :] / temperature
        probs = F.softmax(logits, dim=-1)
        idx_next = torch.multinomial(probs, num_samples=1)
        idx = torch.cat((idx, idx_next), dim=1)

    generated = sp.decode(idx[0].tolist())
    return generated

# ---------------------------------------------------------
# 5. TEST GENERATION
# ---------------------------------------------------------
test_prompts = [
    "தமிழ் மொழி",
    "இன்று காலை",
    "இந்தியா",
    "காலை வணக்கம்",
    "தமிழ் நாடு"
]

print("\n" + "="*50)
print("🤖 TAMIL LLM - TEXT GENERATION")
print("="*50)

for prompt in test_prompts:
    output = generate_text(prompt, max_new_tokens=30, temperature=0.8)
    print(f"🔹 Generated: {output}\n")
```

```python
"""
Tamil LLM v2 - உண்மையான உரை உருவாக்கம்
=========================================
Fixes from v1:
  1. Diverse corpus (no simple repetition)
  2. Train/Val split with overfitting detection
  3. Top-p (nucleus) + Top-k sampling
  4. Learning rate scheduler (cosine)
  5. Gradient clipping
  6. Temperature scaling works properly
  7. Larger model config options
  8. Early stopping
"""

import os
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, random_split
import sentencepiece as spm
import math

# =========================================================
# 0. CORPUS - பல்வேறு வகையான தமிழ் வாக்கியங்கள்
#    (உண்மையான தரவு பயன்படுத்த: CC-100 Tamil / IndicCorp)
# =========================================================
TAMIL_CORPUS = """
தமிழ் மொழி மிகவும் பழமையான மொழி.
இது இந்தியாவில் பேசப்படும் முக்கிய மொழிகளில் ஒன்று.
தமிழ் இலக்கியம் பல்லாயிரம் ஆண்டுகள் பழமையானது.
சங்க கால இலக்கியங்கள் தமிழின் சிறப்பை காட்டுகின்றன.
தமிழ் மக்கள் தங்கள் மொழியை பெருமையாக கருதுகின்றனர்.
உலகின் பல பகுதிகளிலும் தமிழ் பேசப்படுகிறது.
தமிழ் நாட்டில் பல அழகான இடங்கள் உள்ளன.
சென்னை தமிழ் நாட்டின் தலைநகரம் ஆகும்.
மதுரை மிகவும் பழமையான நகரம்.
திருச்சி கோட்டை மிகவும் பிரசித்தமானது.
கன்னியாகுமரி இந்தியாவின் தென்கோடியில் அமைந்துள்ளது.
ஊட்டி மலை நகரம் குளிரான வானிலைக்கு பெயர் பெற்றது.
தமிழ் உணவுகள் மிகவும் சுவையானவை.
இட்லி தோசை தமிழ் மக்களின் விருப்பமான உணவு.
சாம்பார் ரசம் தமிழ் சமையலின் சிறப்பு.
பொங்கல் திருவிழா மிகவும் சிறப்பானது.
தைப்பொங்கல் தமிழ் மக்களின் முக்கிய திருவிழா.
காவிரி நதி தமிழ் நாட்டின் முக்கிய நதி.
காவிரி கரையில் பல கோவில்கள் உள்ளன.
மீனாட்சி அம்மன் கோவில் உலகப் புகழ் பெற்றது.
பழனி முருகன் கோவில் மிகவும் புகழ் பெற்றது.
திருவண்ணாமலை அருணாசலேஸ்வரர் கோவில் பிரசித்தம்.
தமிழ் கல்வி மிக முக்கியமானது.
கல்வி கற்பது ஒவ்வொருவரின் உரிமை.
தமிழ் மொழி கற்றால் உலகமே வெல்லலாம்.
அறிவியல் படிப்போம் நாட்டை வளர்ப்போம்.
தமிழ் இசை மிகவும் இனிமையானது.
கர்நாடக இசை தமிழ் நாட்டின் பாரம்பரியம்.
திரையிசை தமிழ் மக்களை மகிழ்விக்கிறது.
ஏ ஆர் ரஹ்மான் உலகப் புகழ் பெற்ற இசையமைப்பாளர்.
தமிழ் திரைப்படங்கள் உலகம் முழுவதும் பார்க்கப்படுகின்றன.
விவசாயம் தமிழ் நாட்டின் முக்கிய தொழில்.
நெல் கோதுமை தமிழ் நாட்டின் முக்கிய பயிர்கள்.
வாழை மா தென்னை தமிழ் நாட்டில் அதிகம் விளைகின்றன.
மீன்பிடித் தொழில் கடலோர மக்களின் வாழ்வாதாரம்.
தமிழ் நாட்டில் தொழிற்சாலைகள் அதிகமாக உள்ளன.
சென்னை தகவல் தொழில்நுட்ப நகரமாக வளர்ந்துள்ளது.
கோயம்புத்தூர் தொழில் நகரம் என்று அழைக்கப்படுகிறது.
திருப்பூர் துணி உற்பத்திக்கு பிரசித்தம்.
தமிழர்கள் கணிதம் அறிவியலில் சிறந்தவர்கள்.
ராமானுஜன் உலகப் புகழ் பெற்ற கணிதமேதை.
அப்துல் கலாம் தமிழ் மண்ணில் பிறந்த மாமேதை.
தமிழ் நாட்டு மக்கள் கல்விக்கு அதிக முக்கியத்துவம் தருகிறார்கள்.
புத்தகங்கள் படிப்பது மனதை வளப்படுத்தும்.
நூலகங்கள் அறிவின் கோட்டைகள்.
வாசிப்பு பழக்கம் நல்ல குணத்தை வளர்க்கும்.
தமிழ் கவிதைகள் இதயத்தை தொடும்.
சுப்பிரமணிய பாரதி தமிழின் சிறந்த கவிஞர்.
திருவள்ளுவர் குறள் மானிடர்க்கு வழிகாட்டி.
திருக்குறள் உலகின் சிறந்த நீதி நூல்.
அறம் செய விரும்பு ஆவது மனிதன்.
கல்லாதது கைமண்ணளவு கற்றது உலகளவு.
யாதும் ஊரே யாவரும் கேளிர்.
வாழ்க்கை ஒரு பயணம் அதை மகிழ்வாக கழிப்போம்.
நண்பர்கள் வாழ்க்கையில் மிக முக்கியமானவர்கள்.
குடும்பம் நம் வாழ்வின் அடிப்படை.
அன்பு வாழ்க்கையை அழகாக்கும்.
உழைப்பே உயர்வுக்கு வழி.
நேர்மை சிறந்த பண்பு.
தன்னம்பிக்கை வெற்றிக்கு வழி.
கனவு காண்பவனே உயர்வு அடைவான்.
இன்று செய்யும் உழைப்பு நாளைய வெற்றி.
தமிழ் நாட்டில் கல்வி மிகவும் வளர்ந்துள்ளது.
பள்ளிகள் கல்லூரிகள் ஆராய்ச்சி நிலையங்கள் பெருகியுள்ளன.
மாணவர்கள் கடுமையாக படித்து வெற்றி பெறுகிறார்கள்.
ஆசிரியர்கள் நாட்டின் தூண்கள்.
மருத்துவர்கள் மனிதர்களுக்கு கடவுளர்கள்.
விஞ்ஞானிகள் உலகை மாற்றுகிறார்கள்.
பொறியாளர்கள் நாட்டை கட்டுகிறார்கள்.
விவசாயிகள் உலகுக்கு உணவு தருகிறார்கள்.
தமிழ் கலாச்சாரம் மிகவும் பழமையானது.
கோலம் போடுவது தமிழ் பண்பாட்டின் அடையாளம்.
சிலம்பாட்டம் தமிழரின் போர்க்கலை.
தமிழ் நாட்டு நடனம் பரதநாட்டியம் உலகப் புகழ் பெற்றது.
தஞ்சாவூர் ஓவியம் தமிழ் கலையின் சிகரம்.
செம்மொழி தமிழ் என்றும் வாழும்.
"""

def build_diverse_corpus(base_text: str, output_path: str) -> None:
    """
    உண்மையான diversity உருவாக்க corpus-ஐ augment செய்கிறோம்.
    Simple repetition-ஐ தவிர்க்கிறோம் - வேறு வழிகளில் augment செய்கிறோம்.
    """
    sentences = [s.strip() for s in base_text.strip().split('\n') if s.strip()]

    augmented = []
    # Pass 1: அசல் வாக்கியங்கள்
    augmented.extend(sentences)
    # Pass 2: இரண்டு வாக்கியங்களை சேர்த்து எழுதுதல்
    for i in range(len(sentences) - 1):
        augmented.append(sentences[i] + ' ' + sentences[i + 1])
    # Pass 3: மூன்று வாக்கியங்களை சேர்த்து எழுதுதல்
    for i in range(len(sentences) - 2):
        augmented.append(sentences[i] + ' ' + sentences[i + 1] + ' ' + sentences[i + 2])
    # Pass 4: Random shuffled paragraphs (5 வாக்கியம் தொகுதிகளாக)
    import random
    random.seed(42)
    for _ in range(20):
        sample = random.sample(sentences, min(5, len(sentences)))
        augmented.append(' '.join(sample))

    full_text = '\n'.join(augmented)
    with open(output_path, 'w', encoding='utf-8') as f:
        f.write(full_text)

    print(f"✅ Corpus: {len(sentences)} அசல் வாக்கியங்கள் → {len(augmented)} augmented samples")
    print(f"📏 மொத்த characters: {len(full_text)}")


# =========================================================
# 1. CONFIGURATION
# =========================================================
config = {
    # Data
    "corpus_path":     "/kaggle/working/tamil_corpus_v2.txt",
    "tokenizer_path":  "/kaggle/working/tamil_tok_v2",
    "model_save_path": "/kaggle/working/tamil_llm_v2.pth",
    "vocab_size":      600,       # v1 ஐவிட அதிகம்
    # Model
    "block_size":      64,        # நீண்ட context
    "n_embd":          128,       # அதிக embedding dims
    "n_head":          4,
    "n_layer":         4,
    "dropout":         0.15,
    # Training
    "batch_size":      8,
    "learning_rate":   3e-4,
    "epochs":          80,
    "val_split":       0.1,       # 10% validation
    "patience":        10,        # Early stopping
    "grad_clip":       1.0,       # Gradient clipping
}

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"💻 Device: {device}")


# =========================================================
# 2. BUILD CORPUS & TOKENIZER
# =========================================================
build_diverse_corpus(TAMIL_CORPUS, config["corpus_path"])

spm.SentencePieceTrainer.train(
    input=config["corpus_path"],
    model_prefix=config["tokenizer_path"],
    vocab_size=config["vocab_size"],
    character_coverage=1.0,
    model_type='bpe',
    num_threads=1,
    shuffle_input_sentence=True,
    # Tamil-specific: word boundaries பயன்படுத்துதல்
    split_by_unicode_script=True,
    split_by_whitespace=True,
)
print("✅ Tokenizer v2 தயார்!")


# =========================================================
# 3. DATASET WITH TRAIN/VAL SPLIT
# =========================================================
class TamilDataset(Dataset):
    def __init__(self, tokens: list, block_size: int):
        self.tokens = tokens
        self.block_size = block_size

    def __len__(self):
        return max(0, len(self.tokens) - self.block_size - 1)

    def __getitem__(self, idx):
        x = torch.tensor(self.tokens[idx     : idx + self.block_size],     dtype=torch.long)
        y = torch.tensor(self.tokens[idx + 1 : idx + self.block_size + 1], dtype=torch.long)
        return x, y


sp = spm.SentencePieceProcessor()
sp.load(config["tokenizer_path"] + ".model")

with open(config["corpus_path"], 'r', encoding='utf-8') as f:
    raw_text = f.read()

all_tokens = sp.encode(raw_text, out_type=int)
print(f"📊 Total tokens: {len(all_tokens)}")

# Train / Val split (token level - data leakage தவிர்க்க)
split_at = int(len(all_tokens) * (1 - config["val_split"]))
train_tokens = all_tokens[:split_at]
val_tokens   = all_tokens[split_at:]

train_dataset = TamilDataset(train_tokens, config["block_size"])
val_dataset   = TamilDataset(val_tokens,   config["block_size"])

train_loader = DataLoader(train_dataset, batch_size=config["batch_size"], shuffle=True,  drop_last=True)
val_loader   = DataLoader(val_dataset,   batch_size=config["batch_size"], shuffle=False, drop_last=True)

print(f"🔀 Train samples: {len(train_dataset)}, Val samples: {len(val_dataset)}")


# =========================================================
# 4. MODEL ARCHITECTURE (v1 ஐவிட ஆழமான)
# =========================================================
class CausalSelfAttention(nn.Module):
    """Multi-head causal self-attention with Flash-style efficiency."""
    def __init__(self):
        super().__init__()
        assert config["n_embd"] % config["n_head"] == 0
        self.n_head  = config["n_head"]
        self.n_embd  = config["n_embd"]
        self.head_sz = config["n_embd"] // config["n_head"]

        self.c_attn  = nn.Linear(config["n_embd"], 3 * config["n_embd"], bias=False)
        self.c_proj  = nn.Linear(config["n_embd"], config["n_embd"],     bias=False)
        self.drop_attn = nn.Dropout(config["dropout"])
        self.drop_proj = nn.Dropout(config["dropout"])

        self.register_buffer(
            'mask',
            torch.tril(torch.ones(config["block_size"], config["block_size"]))
            .view(1, 1, config["block_size"], config["block_size"])
        )

    def forward(self, x):
        B, T, C = x.shape
        # Q K V ஒரே linear-ல் கணக்கிட்டு பிரிக்கிறோம்
        q, k, v = self.c_attn(x).split(self.n_embd, dim=2)

        # heads-ஆக மாற்றுகிறோம்
        q = q.view(B, T, self.n_head, self.head_sz).transpose(1, 2)  # (B,H,T,Hs)
        k = k.view(B, T, self.n_head, self.head_sz).transpose(1, 2)
        v = v.view(B, T, self.n_head, self.head_sz).transpose(1, 2)

        # Scaled dot-product attention
        scale = math.sqrt(self.head_sz)
        att = (q @ k.transpose(-2, -1)) / scale
        att = att.masked_fill(self.mask[:, :, :T, :T] == 0, float('-inf'))
        att = F.softmax(att, dim=-1)
        att = self.drop_attn(att)

        out = att @ v
        out = out.transpose(1, 2).contiguous().view(B, T, C)
        out = self.drop_proj(self.c_proj(out))
        return out


class MLP(nn.Module):
    """GELU-based MLP (v1 ReLU ஐவிட சிறந்தது)."""
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(config["n_embd"], 4 * config["n_embd"]),
            nn.GELU(),                          # ReLU → GELU (better gradients)
            nn.Linear(4 * config["n_embd"], config["n_embd"]),
            nn.Dropout(config["dropout"]),
        )
    def forward(self, x):
        return self.net(x)


class TransformerBlock(nn.Module):
    def __init__(self):
        super().__init__()
        self.ln1 = nn.LayerNorm(config["n_embd"])
        self.ln2 = nn.LayerNorm(config["n_embd"])
        self.attn = CausalSelfAttention()
        self.mlp  = MLP()

    def forward(self, x):
        x = x + self.attn(self.ln1(x))  # Pre-norm (v1 இல் post-norm இருந்தது)
        x = x + self.mlp(self.ln2(x))
        return x


class TamilLLM(nn.Module):
    def __init__(self, vocab_size: int):
        super().__init__()
        self.tok_emb = nn.Embedding(vocab_size, config["n_embd"])
        self.pos_emb = nn.Embedding(config["block_size"], config["n_embd"])
        self.drop    = nn.Dropout(config["dropout"])
        self.blocks  = nn.Sequential(*[TransformerBlock() for _ in range(config["n_layer"])])
        self.ln_f    = nn.LayerNorm(config["n_embd"])
        self.lm_head = nn.Linear(config["n_embd"], vocab_size, bias=False)

        # Weight tying (embedding ↔ lm_head) - parameter efficiency
        self.tok_emb.weight = self.lm_head.weight

        # Initialize weights properly
        self.apply(self._init_weights)

    def _init_weights(self, module):
        if isinstance(module, nn.Linear):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)
            if module.bias is not None:
                nn.init.zeros_(module.bias)
        elif isinstance(module, nn.Embedding):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)

    def forward(self, idx, targets=None):
        B, T = idx.shape
        pos  = torch.arange(T, device=idx.device)

        x = self.drop(self.tok_emb(idx) + self.pos_emb(pos))
        x = self.blocks(x)
        x = self.ln_f(x)
        logits = self.lm_head(x)

        loss = None
        if targets is not None:
            loss = F.cross_entropy(logits.view(-1, logits.size(-1)), targets.view(-1))
        return logits, loss

    @torch.no_grad()
    def generate(self, idx, max_new_tokens: int, temperature: float = 1.0,
                 top_k: int = 50, top_p: float = 0.9) -> torch.Tensor:
        """
        Top-p (nucleus) + Top-k sampling
        v1 ஐவிட மிகவும் மேம்படுத்தப்பட்ட text generation.
        """
        for _ in range(max_new_tokens):
            idx_cond = idx[:, -config["block_size"]:]
            logits, _ = self(idx_cond)
            logits = logits[:, -1, :] / max(temperature, 1e-8)

            # Top-k filtering
            if top_k > 0:
                top_k_vals, _ = torch.topk(logits, min(top_k, logits.size(-1)))
                logits[logits < top_k_vals[:, -1:]] = float('-inf')

            # Top-p (nucleus) filtering
            if 0.0 < top_p < 1.0:
                sorted_logits, sorted_idx = torch.sort(logits, descending=True)
                cumul_probs = torch.cumsum(F.softmax(sorted_logits, dim=-1), dim=-1)
                # Cumulative threshold-ஐ தாண்டிய tokens அகற்றுதல்
                remove_mask = cumul_probs - F.softmax(sorted_logits, dim=-1) > top_p
                sorted_logits[remove_mask] = float('-inf')
                logits = logits.scatter(1, sorted_idx, sorted_logits)

            probs = F.softmax(logits, dim=-1)
            next_tok = torch.multinomial(probs, num_samples=1)
            idx = torch.cat([idx, next_tok], dim=1)

        return idx


# =========================================================
# 5. TRAINING LOOP WITH VAL LOSS + EARLY STOPPING
# =========================================================
vocab_size = sp.get_piece_size()
model      = TamilLLM(vocab_size).to(device)
optimizer  = torch.optim.AdamW(
    model.parameters(),
    lr=config["learning_rate"],
    weight_decay=0.1,   # L2 regularization
    betas=(0.9, 0.95),
)

# Cosine LR scheduler (v1 இல் scheduler இல்லை)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=config["epochs"],
    eta_min=config["learning_rate"] / 10,
)

n_params = sum(p.numel() for p in model.parameters()) / 1e6
print(f"\n🔧 Parameters: {n_params:.2f}M")
print(f"📊 Vocab size: {vocab_size}")
print(f"🔁 Epochs: {config['epochs']} (patience={config['patience']})\n")


@torch.no_grad()
def evaluate(loader) -> float:
    model.eval()
    total_loss, n = 0.0, 0
    for x, y in loader:
        x, y = x.to(device), y.to(device)
        _, loss = model(x, y)
        total_loss += loss.item()
        n += 1
    return total_loss / max(n, 1)


history = {"train": [], "val": []}
best_val_loss = float('inf')
patience_counter = 0

for epoch in range(1, config["epochs"] + 1):
    # --- Training ---
    model.train()
    train_loss, n = 0.0, 0

    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        _, loss = model(x, y)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        # Gradient clipping: exploding gradient தடுக்கிறோம்
        torch.nn.utils.clip_grad_norm_(model.parameters(), config["grad_clip"])
        optimizer.step()

        train_loss += loss.item()
        n += 1

    scheduler.step()

    avg_train = train_loss / max(n, 1)
    avg_val   = evaluate(val_loader)
    history["train"].append(avg_train)
    history["val"].append(avg_val)

    current_lr = scheduler.get_last_lr()[0]
    print(f"Epoch {epoch:3d}/{config['epochs']} | "
          f"Train: {avg_train:.4f} | Val: {avg_val:.4f} | LR: {current_lr:.6f}")

    # Best model சேமித்தல்
    if avg_val < best_val_loss:
        best_val_loss = avg_val
        patience_counter = 0
        torch.save({
            "model_state": model.state_dict(),
            "config":      config,
            "vocab_size":  vocab_size,
            "epoch":       epoch,
            "val_loss":    best_val_loss,
        }, config["model_save_path"])
        print(f"    💾 Best model saved! Val loss: {best_val_loss:.4f}")
    else:
        patience_counter += 1
        if patience_counter >= config["patience"]:
            print(f"\n⏹️  Early stopping at epoch {epoch} (no improvement for {config['patience']} epochs)")
            break

print(f"\n✅ Training complete! Best val loss: {best_val_loss:.4f}")


# =========================================================
# 6. LOAD BEST MODEL & GENERATE
# =========================================================
checkpoint = torch.load(config["model_save_path"], map_location=device)
model.load_state_dict(checkpoint["model_state"])
model.eval()
print(f"\n✅ Best model loaded (Epoch {checkpoint['epoch']}, Val={checkpoint['val_loss']:.4f})")


def generate_text(prompt: str, max_tokens: int = 60,
                  temperature: float = 0.8, top_k: int = 40, top_p: float = 0.9) -> str:
    """உரை உருவாக்கும் function."""
    tokens = sp.encode(prompt, out_type=int)
    idx    = torch.tensor([tokens], dtype=torch.long, device=device)
    out    = model.generate(idx, max_tokens, temperature=temperature, top_k=top_k, top_p=top_p)
    return sp.decode(out[0].tolist())


# =========================================================
# 7. TEST GENERATION - வெவ்வேறு temperature settings
# =========================================================
print("\n" + "=" * 60)
print("🤖  TAMIL LLM v2 - TEXT GENERATION")
print("=" * 60)

test_cases = [
    # (prompt, temperature, top_k, top_p)
    ("தமிழ் மொழி",     0.8, 40, 0.9),
    ("தமிழ் மொழி",     1.2, 50, 0.95),   # அதிக creativity
    ("இன்று காலை",     0.6, 20, 0.8),    # அதிக focused
    ("இந்தியா",        0.9, 40, 0.9),
    ("கல்வி என்பது",   0.8, 40, 0.9),
    ("திருவள்ளுவர்",   0.8, 40, 0.9),
    ("தமிழ் உணவு",     0.7, 30, 0.85),
]

for prompt, temp, tk, tp in test_cases:
    output = generate_text(prompt, max_tokens=50, temperature=temp, top_k=tk, top_p=tp)
    print(f"\n📝 Prompt : '{prompt}' (temp={temp}, top_k={tk}, top_p={tp})")
    print(f"🔹 Output : {output}")
    print("-" * 60)

# =========================================================
# 8. COMPARE DIVERSITY: generate same prompt 3 times
# =========================================================
print("\n\n🎲 DIVERSITY TEST - ஒரே prompt, 3 மாறுபட்ட outputs:")
print("=" * 60)
for i in range(3):
    out = generate_text("தமிழ் மொழி", max_tokens=40, temperature=1.0, top_k=50, top_p=0.95)
    print(f"  {i+1}. {out}")
```

```python
"""
Tamila - Tamil LLM v3
=====================
Fixes from v2:
  - Repetition loops fixed  (repetition_penalty + n-gram blocking)
  - Overfitting fixed        (smaller model, dropout=0.30, bigger corpus)
  - Train/Val gap target     < 3x  (v2 was 10x)
"""

import os, math, random
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import sentencepiece as spm

# =========================================================
# 0. CORPUS
# =========================================================
CORPUS_SENTENCES = [
    "தமிழ் மொழி மிகவும் பழமையான மொழி.",
    "தமிழ் உலகின் செம்மொழிகளில் ஒன்று.",
    "தமிழ் இலக்கியம் பல்லாயிரம் ஆண்டுகள் பழமையானது.",
    "சங்க கால இலக்கியங்கள் தமிழின் சிறப்பை காட்டுகின்றன.",
    "தமிழ் மக்கள் தங்கள் மொழியை பெருமையாக கருதுகின்றனர்.",
    "உலகின் பல பகுதிகளிலும் தமிழ் பேசப்படுகிறது.",
    "தமிழ் மொழி இனிமையானது இசையானது.",
    "திருக்குறள் உலகின் சிறந்த நீதி நூல்.",
    "திருவள்ளுவர் குறள் மானிடர்க்கு வழிகாட்டி.",
    "சுப்பிரமணிய பாரதி தமிழின் சிறந்த கவிஞர்.",
    "யாதும் ஊரே யாவரும் கேளிர் என்பது தமிழ் தத்துவம்.",
    "அறம் செய விரும்பு ஆவது மனிதன்.",
    "கல்லாதது கைமண்ணளவு கற்றது உலகளவு.",
    "இலக்கணம் தமிழின் அடிப்படை.",
    "சென்னை தமிழ் நாட்டின் தலைநகரம் ஆகும்.",
    "மதுரை மிகவும் பழமையான நகரம்.",
    "திருச்சி கோட்டை மிகவும் பிரசித்தமானது.",
    "கன்னியாகுமரி இந்தியாவின் தென்கோடியில் அமைந்துள்ளது.",
    "ஊட்டி மலை நகரம் குளிரான வானிலைக்கு பெயர் பெற்றது.",
    "காவிரி நதி தமிழ் நாட்டின் முக்கிய நதி.",
    "காவிரி கரையில் பல கோவில்கள் உள்ளன.",
    "மீனாட்சி அம்மன் கோவில் உலகப் புகழ் பெற்றது.",
    "பழனி முருகன் கோவில் மிகவும் புகழ் பெற்றது.",
    "திருவண்ணாமலை அருணாசலேஸ்வரர் கோவில் பிரசித்தம்.",
    "தஞ்சாவூர் பெரிய கோவில் தமிழ் கட்டிடக்கலையின் சிகரம்.",
    "ராமேஸ்வரம் புனித தீர்த்தத்தலம் ஆகும்.",
    "கல்வி கற்பது ஒவ்வொருவரின் உரிமை.",
    "தமிழ் மொழி கற்றால் உலகமே வெல்லலாம்.",
    "அறிவியல் படிப்போம் நாட்டை வளர்ப்போம்.",
    "புத்தகங்கள் படிப்பது மனதை வளப்படுத்தும்.",
    "நூலகங்கள் அறிவின் கோட்டைகள்.",
    "வாசிப்பு பழக்கம் நல்ல குணத்தை வளர்க்கும்.",
    "ராமானுஜன் உலகப் புகழ் பெற்ற கணிதமேதை.",
    "அப்துல் கலாம் தமிழ் மண்ணில் பிறந்த மாமேதை.",
    "மாணவர்கள் கடுமையாக படித்து வெற்றி பெறுகிறார்கள்.",
    "ஆசிரியர்கள் நாட்டின் தூண்கள்.",
    "வாழ்க்கை ஒரு பயணம் அதை மகிழ்வாக கழிப்போம்.",
    "நண்பர்கள் வாழ்க்கையில் மிக முக்கியமானவர்கள்.",
    "குடும்பம் நம் வாழ்வின் அடிப்படை.",
    "அன்பு வாழ்க்கையை அழகாக்கும்.",
    "உழைப்பே உயர்வுக்கு வழி.",
    "நேர்மை சிறந்த பண்பு.",
    "தன்னம்பிக்கை வெற்றிக்கு வழி.",
    "கனவு காண்பவனே உயர்வு அடைவான்.",
    "இன்று செய்யும் உழைப்பு நாளைய வெற்றி.",
    "பொறுமை எல்லாவற்றையும் வெல்லும்.",
    "அன்பே சிவம் என்பது தமிழரின் நம்பிக்கை.",
    "இட்லி தோசை தமிழ் மக்களின் விருப்பமான உணவு.",
    "சாம்பார் ரசம் தமிழ் சமையலின் சிறப்பு.",
    "பொங்கல் திருவிழா மிகவும் சிறப்பானது.",
    "தைப்பொங்கல் தமிழ் மக்களின் முக்கிய திருவிழா.",
    "தமிழ் உணவுகள் மிகவும் சுவையானவை.",
    "வாழை மா தென்னை தமிழ் நாட்டில் அதிகம் விளைகின்றன.",
    "தமிழ் இசை மிகவும் இனிமையானது.",
    "கர்நாடக இசை தமிழ் நாட்டின் பாரம்பரியம்.",
    "திரையிசை தமிழ் மக்களை மகிழ்விக்கிறது.",
    "ஏ ஆர் ரஹ்மான் உலகப் புகழ் பெற்ற இசையமைப்பாளர்.",
    "தமிழ் திரைப்படங்கள் உலகம் முழுவதும் பார்க்கப்படுகின்றன.",
    "பரதநாட்டியம் தமிழரின் செவ்வியல் நடனம்.",
    "தஞ்சாவூர் ஓவியம் தமிழ் கலையின் அடையாளம்.",
    "சிலம்பாட்டம் தமிழரின் போர்க்கலை.",
    "கோலம் போடுவது தமிழ் பண்பாட்டின் அடையாளம்.",
    "விவசாயம் தமிழ் நாட்டின் முக்கிய தொழில்.",
    "நெல் கோதுமை தமிழ் நாட்டின் முக்கிய பயிர்கள்.",
    "மீன்பிடித் தொழில் கடலோர மக்களின் வாழ்வாதாரம்.",
    "சென்னை தகவல் தொழில்நுட்ப நகரமாக வளர்ந்துள்ளது.",
    "கோயம்புத்தூர் தொழில் நகரம் என்று அழைக்கப்படுகிறது.",
    "திருப்பூர் துணி உற்பத்திக்கு பிரசித்தம்.",
    "மருத்துவர்கள் மனிதர்களுக்கு கடவுளர்கள்.",
    "விஞ்ஞானிகள் உலகை மாற்றுகிறார்கள்.",
    "பொறியாளர்கள் நாட்டை கட்டுகிறார்கள்.",
    "விவசாயிகள் உலகுக்கு உணவு தருகிறார்கள்.",
]


def build_corpus(sentences, output_path):
    random.seed(42)
    samples = list(sentences)

    # Adjacent pairs (gap 1 and 2)
    for gap in [1, 2]:
        for i in range(len(sentences) - gap):
            samples.append(sentences[i] + " " + sentences[i + gap])

    # Triplets
    for i in range(len(sentences) - 2):
        samples.append(sentences[i] + " " + sentences[i+1] + " " + sentences[i+2])

    # Random paragraphs (5-7 sentences)
    for _ in range(60):
        k = random.randint(5, 7)
        samples.append(" ".join(random.sample(sentences, k)))

    # Reversed pairs
    for i in range(len(sentences) - 1):
        samples.append(sentences[i+1] + " " + sentences[i])

    # Mid-join
    for i in range(len(sentences) - 1):
        words = sentences[i].split()
        if len(words) > 3:
            half = " ".join(words[len(words)//2:])
            samples.append(half + " " + sentences[i+1])

    random.shuffle(samples)
    text = "\n".join(samples)
    with open(output_path, "w", encoding="utf-8") as f:
        f.write(text)
    print("[OK] Corpus: " + str(len(sentences)) + " sentences -> " + str(len(samples)) + " unique samples")
    print("[INFO] Characters: " + str(len(text)))


# =========================================================
# 1. CONFIG
# =========================================================
config = {
    "corpus_path":     "/kaggle/working/tamila_corpus.txt",
    "tokenizer_path":  "/kaggle/working/tamila_tok",
    "model_save_path": "/kaggle/working/tamila_model.pth",
    "vocab_size":      800,
    "block_size":      48,
    "n_embd":          96,
    "n_head":          4,
    "n_layer":         3,
    "dropout":         0.30,
    "batch_size":      16,
    "learning_rate":   2e-4,
    "min_lr":          2e-5,
    "epochs":          100,
    "val_split":       0.15,
    "patience":        15,
    "grad_clip":       0.8,
    "weight_decay":    0.15,
}

device = "cuda" if torch.cuda.is_available() else "cpu"
print("[INFO] Device: " + device)

# =========================================================
# 2. BUILD CORPUS + TOKENIZER
# =========================================================
build_corpus(CORPUS_SENTENCES, config["corpus_path"])

spm.SentencePieceTrainer.train(
    input=config["corpus_path"],
    model_prefix=config["tokenizer_path"],
    vocab_size=config["vocab_size"],
    character_coverage=1.0,
    model_type="bpe",
    num_threads=1,
    shuffle_input_sentence=True,
)
print("[OK] Tokenizer ready!")

sp = spm.SentencePieceProcessor()
sp.load(config["tokenizer_path"] + ".model")

# =========================================================
# 3. DATASET
# =========================================================
class TamilDataset(Dataset):
    def __init__(self, tokens, block_size):
        self.tokens = tokens
        self.block_size = block_size

    def __len__(self):
        return max(0, len(self.tokens) - self.block_size - 1)

    def __getitem__(self, idx):
        chunk = self.tokens[idx : idx + self.block_size + 1]
        x = torch.tensor(chunk[:-1], dtype=torch.long)
        y = torch.tensor(chunk[1:],  dtype=torch.long)
        return x, y


with open(config["corpus_path"], "r", encoding="utf-8") as f:
    all_tokens = sp.encode(f.read(), out_type=int)

split = int(len(all_tokens) * (1 - config["val_split"]))
train_ds = TamilDataset(all_tokens[:split], config["block_size"])
val_ds   = TamilDataset(all_tokens[split:], config["block_size"])

train_loader = DataLoader(train_ds, batch_size=config["batch_size"], shuffle=True,  drop_last=True)
val_loader   = DataLoader(val_ds,   batch_size=config["batch_size"], shuffle=False, drop_last=True)

print("[INFO] Tokens: " + str(len(all_tokens)) +
      " | Train samples: " + str(len(train_ds)) +
      " | Val samples: "   + str(len(val_ds)))

# =========================================================
# 4. MODEL
# =========================================================
class CausalSelfAttention(nn.Module):
    def __init__(self):
        super().__init__()
        self.n_head  = config["n_head"]
        self.n_embd  = config["n_embd"]
        self.head_sz = self.n_embd // self.n_head
        self.c_attn  = nn.Linear(self.n_embd, 3 * self.n_embd, bias=False)
        self.c_proj  = nn.Linear(self.n_embd, self.n_embd, bias=False)
        self.attn_drop = nn.Dropout(config["dropout"])
        self.proj_drop = nn.Dropout(config["dropout"])
        self.register_buffer(
            "mask",
            torch.tril(torch.ones(config["block_size"], config["block_size"]))
            .view(1, 1, config["block_size"], config["block_size"])
        )

    def forward(self, x):
        B, T, C = x.shape
        q, k, v = self.c_attn(x).split(self.n_embd, dim=2)
        def rs(t):
            return t.view(B, T, self.n_head, self.head_sz).transpose(1, 2)
        q, k, v = rs(q), rs(k), rs(v)
        att = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_sz)
        att = att.masked_fill(self.mask[:, :, :T, :T] == 0, float("-inf"))
        att = self.attn_drop(F.softmax(att, dim=-1))
        out = (att @ v).transpose(1, 2).contiguous().view(B, T, C)
        return self.proj_drop(self.c_proj(out))


class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(config["n_embd"], 4 * config["n_embd"]),
            nn.GELU(),
            nn.Linear(4 * config["n_embd"], config["n_embd"]),
            nn.Dropout(config["dropout"]),
        )
    def forward(self, x):
        return self.net(x)


class Block(nn.Module):
    def __init__(self):
        super().__init__()
        self.ln1  = nn.LayerNorm(config["n_embd"])
        self.ln2  = nn.LayerNorm(config["n_embd"])
        self.attn = CausalSelfAttention()
        self.mlp  = MLP()

    def forward(self, x):
        x = x + self.attn(self.ln1(x))
        x = x + self.mlp(self.ln2(x))
        return x


class TamilLLM(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.tok_emb = nn.Embedding(vocab_size, config["n_embd"])
        self.pos_emb = nn.Embedding(config["block_size"], config["n_embd"])
        self.drop    = nn.Dropout(config["dropout"])
        self.blocks  = nn.Sequential(*[Block() for _ in range(config["n_layer"])])
        self.ln_f    = nn.LayerNorm(config["n_embd"])
        self.lm_head = nn.Linear(config["n_embd"], vocab_size, bias=False)
        self.tok_emb.weight = self.lm_head.weight  # weight tying
        self.apply(self._init_weights)

    def _init_weights(self, m):
        if isinstance(m, nn.Linear):
            nn.init.normal_(m.weight, 0.0, 0.02)
            if m.bias is not None:
                nn.init.zeros_(m.bias)
        elif isinstance(m, nn.Embedding):
            nn.init.normal_(m.weight, 0.0, 0.02)

    def forward(self, idx, targets=None):
        B, T = idx.shape
        x = self.drop(
            self.tok_emb(idx) + self.pos_emb(torch.arange(T, device=idx.device))
        )
        x = self.ln_f(self.blocks(x))
        logits = self.lm_head(x)
        loss = None
        if targets is not None:
            loss = F.cross_entropy(logits.view(-1, logits.size(-1)), targets.view(-1))
        return logits, loss

    @torch.no_grad()
    def generate(self, idx, max_new_tokens,
                 temperature=0.9, top_k=50, top_p=0.9,
                 repetition_penalty=1.3, no_repeat_ngram=3):
        for _ in range(max_new_tokens):
            idx_cond = idx[:, -config["block_size"]:]
            logits, _ = self(idx_cond)
            logits = logits[:, -1, :]

            # -- Repetition penalty --
            if repetition_penalty != 1.0:
                for token_id in set(idx[0].tolist()):
                    if logits[0, token_id] > 0:
                        logits[0, token_id] /= repetition_penalty
                    else:
                        logits[0, token_id] *= repetition_penalty

            # -- N-gram blocking --
            if no_repeat_ngram > 1 and idx.shape[1] >= no_repeat_ngram - 1:
                history = idx[0].tolist()
                prefix  = tuple(history[-(no_repeat_ngram - 1):])
                banned  = set()
                for i in range(len(history) - no_repeat_ngram + 1):
                    if tuple(history[i:i + no_repeat_ngram - 1]) == prefix:
                        banned.add(history[i + no_repeat_ngram - 1])
                for tid in banned:
                    logits[0, tid] = float("-inf")

            # -- Temperature --
            logits = logits / max(temperature, 1e-8)

            # -- Top-k --
            if top_k > 0:
                v, _ = torch.topk(logits, min(top_k, logits.size(-1)))
                logits[logits < v[:, -1:]] = float("-inf")

            # -- Top-p --
            if 0.0 < top_p < 1.0:
                sorted_logits, sorted_idx = torch.sort(logits, descending=True)
                cumul = torch.cumsum(F.softmax(sorted_logits, dim=-1), dim=-1)
                remove = cumul - F.softmax(sorted_logits, dim=-1) > top_p
                sorted_logits[remove] = float("-inf")
                logits = logits.scatter(1, sorted_idx, sorted_logits)

            probs    = F.softmax(logits, dim=-1)
            next_tok = torch.multinomial(probs, num_samples=1)
            idx      = torch.cat([idx, next_tok], dim=1)

        return idx


# =========================================================
# 5. TRAINING
# =========================================================
vocab_size = sp.get_piece_size()
model      = TamilLLM(vocab_size).to(device)
optimizer  = torch.optim.AdamW(
    model.parameters(),
    lr=config["learning_rate"],
    weight_decay=config["weight_decay"],
    betas=(0.9, 0.95),
)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer, T_max=config["epochs"], eta_min=config["min_lr"]
)

n_params = sum(p.numel() for p in model.parameters()) / 1e6
print("[INFO] Parameters: " + str(round(n_params, 2)) + "M")
print("[INFO] Vocab size:  " + str(vocab_size))
print("[INFO] Train/Val gap target: < 3x  (v2 was 10x)")
print("")


@torch.no_grad()
def evaluate():
    model.eval()
    total, n = 0.0, 0
    for x, y in val_loader:
        x, y = x.to(device), y.to(device)
        _, loss = model(x, y)
        total += loss.item()
        n += 1
    return total / max(n, 1)


best_val, patience_counter = float("inf"), 0
history = {"train": [], "val": []}

for epoch in range(1, config["epochs"] + 1):
    model.train()
    train_loss, n = 0.0, 0
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        _, loss = model(x, y)
        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), config["grad_clip"])
        optimizer.step()
        train_loss += loss.item()
        n += 1
    scheduler.step()

    avg_train = train_loss / max(n, 1)
    avg_val   = evaluate()
    gap       = avg_val / max(avg_train, 1e-8)
    history["train"].append(avg_train)
    history["val"].append(avg_val)

    print("Epoch " + str(epoch).rjust(3) + "/" + str(config["epochs"]) +
          " | Train: " + str(round(avg_train, 4)) +
          " | Val: "   + str(round(avg_val,   4)) +
          " | Gap: "   + str(round(gap, 1))  + "x" +
          " | LR: "    + str(round(scheduler.get_last_lr()[0], 6)))

    if avg_val < best_val:
        best_val, patience_counter = avg_val, 0
        torch.save({
            "model_state": model.state_dict(),
            "config":      config,
            "vocab_size":  vocab_size,
            "epoch":       epoch,
            "val_loss":    best_val,
        }, config["model_save_path"])
        print("  [SAVED] Best model! Val: " + str(round(best_val, 4)) +
              "  gap=" + str(round(gap, 1)) + "x")
    else:
        patience_counter += 1
        if patience_counter >= config["patience"]:
            print("[STOP] Early stopping at epoch " + str(epoch))
            break

print("[OK] Training complete! Best val: " + str(round(best_val, 4)))

# =========================================================
# 6. LOAD BEST MODEL
# =========================================================
ckpt = torch.load(config["model_save_path"], map_location=device)
model.load_state_dict(ckpt["model_state"])
model.eval()
print("[OK] Best model loaded  epoch=" + str(ckpt["epoch"]) +
      "  val=" + str(round(ckpt["val_loss"], 4)))


def generate(prompt, max_tokens=60, temperature=0.9,
             top_k=50, top_p=0.9, rep_penalty=1.3, no_repeat_ngram=3):
    tokens = sp.encode(prompt, out_type=int)
    idx    = torch.tensor([tokens], dtype=torch.long, device=device)
    out    = model.generate(idx, max_tokens,
                            temperature=temperature,
                            top_k=top_k, top_p=top_p,
                            repetition_penalty=rep_penalty,
                            no_repeat_ngram=no_repeat_ngram)
    return sp.decode(out[0].tolist())


# =========================================================
# 7. GENERATION TESTS
# =========================================================
print("")
print("=" * 60)
print("TAMILA - TEXT GENERATION")
print("=" * 60)

test_cases = [
    ("தமிழ் மொழி",   0.9, 50, 0.90, 1.3, 3),
    ("தமிழ் மொழி",   1.1, 50, 0.95, 1.3, 3),
    ("இன்று காலை",   0.7, 30, 0.85, 1.2, 3),
    ("இந்தியா",      0.9, 40, 0.90, 1.3, 3),
    ("கல்வி என்பது", 0.8, 40, 0.90, 1.3, 3),
    ("திருவள்ளுவர்", 0.8, 40, 0.90, 1.3, 3),
    ("தமிழ் உணவு",   0.8, 40, 0.90, 1.3, 3),
    ("உழைப்பே",      0.9, 50, 0.90, 1.3, 3),
]

for prompt, temp, tk, tp, rep, ngram in test_cases:
    out = generate(prompt, 55, temperature=temp,
                   top_k=tk, top_p=tp,
                   rep_penalty=rep, no_repeat_ngram=ngram)
    print("")
    print("Prompt : " + prompt +
          "  (temp=" + str(temp) + ", rep_penalty=" + str(rep) + ")")
    print("Output : " + out)
    print("-" * 60)

# =========================================================
# 8. DIVERSITY TEST  (3 runs, same prompt - should differ)
# =========================================================
print("")
print("DIVERSITY TEST - same prompt, 3 different outputs:")
print("=" * 60)
for i in range(3):
    out = generate("தமிழ் மொழி", 50,
                   temperature=1.0, top_k=50, top_p=0.95,
                   rep_penalty=1.3, no_repeat_ngram=3)
    print(str(i+1) + ". " + out)

# =========================================================
# 9. FINAL REPORT
# =========================================================
final_train = history["train"][-1]
final_val   = history["val"][-1]
final_gap   = final_val / max(final_train, 1e-8)
print("")
print("OVERFITTING REPORT:")
print("  v2:  train=0.10  val=1.67  gap=16x  [FAIL]")
print("  v3:  train=" + str(round(final_train, 4)) +
      "  val=" + str(round(final_val, 4)) +
      "  gap=" + str(round(final_gap, 1)) + "x" +
      ("  [OK]" if final_gap < 3 else "  [still high - add more corpus data]"))
```

```python
"""
Tamila v4 - Real Tamil NLP Dataset
====================================
Dataset : kagglehub sudalairajkumar/tamil-nlp
          (real Tamil news / articles - NOT repeated sentences)

Changes from v3:
  - Loads real dataset from kagglehub path
  - Cleans Tamil text (removes English, numbers, junk)
  - Auto-detects which .txt/.csv files to use
  - Larger model (more real data = more capacity OK)
  - Perplexity reporting (better metric than raw loss)
"""

import os, re, math, random
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import sentencepiece as spm
import kagglehub

# =========================================================
# 1. DOWNLOAD DATASET
# =========================================================
print("[INFO] Downloading Tamil NLP dataset...")
path = kagglehub.dataset_download("sudalairajkumar/tamil-nlp")
print("[OK] Dataset path: " + path)

# List all files
all_files = []
for root, dirs, files in os.walk(path):
    for fname in files:
        fpath = os.path.join(root, fname)
        all_files.append(fpath)
        print("  Found: " + fpath + "  (" + str(round(os.path.getsize(fpath)/1024, 1)) + " KB)")

# =========================================================
# 2. LOAD + CLEAN TAMIL TEXT
# =========================================================
TAMIL_UNICODE_RANGE = (0x0B80, 0x0BFF)  # Tamil Unicode block

def is_tamil_char(ch):
    return TAMIL_UNICODE_RANGE[0] <= ord(ch) <= TAMIL_UNICODE_RANGE[1]

def clean_tamil_text(text):
    """
    Real dataset-இல் உள்ள junk அகற்றுதல்.
    Tamil characters, spaces, punctuation மட்டும் வைக்கிறோம்.
    """
    lines = text.split("\n")
    clean_lines = []
    for line in lines:
        line = line.strip()
        if not line:
            continue
        # Tamil chars உள்ள lines மட்டும் வைக்கிறோம்
        tamil_count = sum(1 for ch in line if is_tamil_char(ch))
        total_count = len(line.replace(" ", ""))
        if total_count == 0:
            continue
        # At least 50% Tamil characters இருக்க வேண்டும்
        if tamil_count / total_count < 0.5:
            continue
        # English words அகற்றுதல்
        line = re.sub(r'[a-zA-Z0-9]+', ' ', line)
        # Extra spaces சுத்தம்
        line = re.sub(r'\s+', ' ', line).strip()
        # Very short lines தவிர்க்கிறோம்
        if len(line) > 10:
            clean_lines.append(line)
    return "\n".join(clean_lines)


def load_dataset_files(file_list):
    """
    .txt and .csv files-இல் இருந்து Tamil text எடுக்கிறோம்.
    """
    combined_text = ""
    for fpath in file_list:
        ext = os.path.splitext(fpath)[1].lower()
        try:
            if ext == ".txt":
                with open(fpath, "r", encoding="utf-8", errors="ignore") as f:
                    raw = f.read()
                cleaned = clean_tamil_text(raw)
                combined_text += cleaned + "\n"
                print("[OK] Loaded " + os.path.basename(fpath) +
                      " -> " + str(len(cleaned)) + " chars after cleaning")

            elif ext == ".csv":
                with open(fpath, "r", encoding="utf-8", errors="ignore") as f:
                    raw = f.read()
                # CSV-இல் உள்ள Tamil columns எடுக்கிறோம்
                lines = raw.split("\n")
                text_parts = []
                for line in lines[1:]:  # header skip
                    # comma-separated fields-இல் Tamil text தேடுதல்
                    fields = line.split(",")
                    for field in fields:
                        field = field.strip().strip('"')
                        tamil_count = sum(1 for ch in field if is_tamil_char(ch))
                        if tamil_count > 5:
                            text_parts.append(field)
                csv_text = "\n".join(text_parts)
                cleaned = clean_tamil_text(csv_text)
                combined_text += cleaned + "\n"
                print("[OK] Loaded " + os.path.basename(fpath) +
                      " -> " + str(len(cleaned)) + " chars after cleaning")

        except Exception as e:
            print("[WARN] Could not load " + fpath + " : " + str(e))

    return combined_text


print("")
print("[INFO] Loading and cleaning dataset files...")
tamil_text = load_dataset_files(all_files)

if len(tamil_text) < 1000:
    print("[WARN] Very little Tamil text found. Check dataset structure.")
    print("[INFO] Raw file listing:")
    for f in all_files:
        print("  " + f)
else:
    print("")
    print("[OK] Total clean Tamil text: " + str(len(tamil_text)) + " characters")
    print("[INFO] Sample (first 300 chars):")
    print(tamil_text[:300])

# Save cleaned corpus
CORPUS_PATH    = "/kaggle/working/tamila_real_corpus.txt"
TOKENIZER_PATH = "/kaggle/working/tamila_real_tok"
MODEL_PATH     = "/kaggle/working/tamila_real_model.pth"

with open(CORPUS_PATH, "w", encoding="utf-8") as f:
    f.write(tamil_text)
print("")
print("[OK] Corpus saved: " + CORPUS_PATH)

# =========================================================
# 3. CONFIG  (larger model - real data can support it)
# =========================================================
text_size = len(tamil_text)

# Dataset size-க்கு ஏற்ப model size தேர்வு
if text_size > 5_000_000:       # > 5MB
    MODEL_SIZE = "large"
elif text_size > 500_000:       # > 500KB
    MODEL_SIZE = "medium"
else:
    MODEL_SIZE = "small"

print("[INFO] Text size: " + str(text_size) + " chars -> model size: " + MODEL_SIZE)

SIZE_CONFIGS = {
    "small": {
        "vocab_size": 1000, "block_size": 64,
        "n_embd": 128, "n_head": 4, "n_layer": 4,
        "dropout": 0.2, "batch_size": 16,
    },
    "medium": {
        "vocab_size": 4000, "block_size": 128,
        "n_embd": 256, "n_head": 8, "n_layer": 6,
        "dropout": 0.15, "batch_size": 32,
    },
    "large": {
        "vocab_size": 8000, "block_size": 256,
        "n_embd": 384, "n_head": 8, "n_layer": 8,
        "dropout": 0.1, "batch_size": 64,
    },
}

config = {
    **SIZE_CONFIGS[MODEL_SIZE],
    "corpus_path":     CORPUS_PATH,
    "tokenizer_path":  TOKENIZER_PATH,
    "model_save_path": MODEL_PATH,
    "learning_rate":   3e-4,
    "min_lr":          3e-5,
    "epochs":          50,
    "val_split":       0.1,
    "patience":        8,
    "grad_clip":       1.0,
    "weight_decay":    0.1,
}

device = "cuda" if torch.cuda.is_available() else "cpu"
print("[INFO] Device: " + device)

# =========================================================
# 4. TOKENIZER  (real data -> better vocab)
# =========================================================
print("")
print("[INFO] Training tokenizer on real corpus...")
spm.SentencePieceTrainer.train(
    input=config["corpus_path"],
    model_prefix=config["tokenizer_path"],
    vocab_size=config["vocab_size"],
    character_coverage=1.0,
    model_type="bpe",
    num_threads=4,
    shuffle_input_sentence=True,
    # Tamil-specific settings
    split_by_unicode_script=True,
    split_by_whitespace=True,
    # Better coverage of Tamil morphology
    byte_fallback=True,
)
print("[OK] Tokenizer trained! Vocab size: " + str(config["vocab_size"]))

sp = spm.SentencePieceProcessor()
sp.load(config["tokenizer_path"] + ".model")

# =========================================================
# 5. DATASET
# =========================================================
class TamilDataset(Dataset):
    def __init__(self, tokens, block_size):
        self.tokens = tokens
        self.block_size = block_size

    def __len__(self):
        return max(0, len(self.tokens) - self.block_size - 1)

    def __getitem__(self, idx):
        chunk = self.tokens[idx : idx + self.block_size + 1]
        return (torch.tensor(chunk[:-1], dtype=torch.long),
                torch.tensor(chunk[1:],  dtype=torch.long))


print("[INFO] Tokenizing corpus...")
with open(config["corpus_path"], "r", encoding="utf-8") as f:
    raw_text = f.read()

all_tokens = sp.encode(raw_text, out_type=int)
print("[OK] Total tokens: " + str(len(all_tokens)))

split      = int(len(all_tokens) * (1 - config["val_split"]))
train_ds   = TamilDataset(all_tokens[:split], config["block_size"])
val_ds     = TamilDataset(all_tokens[split:], config["block_size"])
train_loader = DataLoader(train_ds, batch_size=config["batch_size"], shuffle=True,  drop_last=True)
val_loader   = DataLoader(val_ds,   batch_size=config["batch_size"], shuffle=False, drop_last=True)

print("[INFO] Train samples: " + str(len(train_ds)) +
      " | Val samples: " + str(len(val_ds)))

# =========================================================
# 6. MODEL  (same architecture, scaled up)
# =========================================================
class CausalSelfAttention(nn.Module):
    def __init__(self):
        super().__init__()
        assert config["n_embd"] % config["n_head"] == 0
        self.n_head  = config["n_head"]
        self.n_embd  = config["n_embd"]
        self.head_sz = self.n_embd // self.n_head
        self.c_attn  = nn.Linear(self.n_embd, 3 * self.n_embd, bias=False)
        self.c_proj  = nn.Linear(self.n_embd, self.n_embd,     bias=False)
        self.attn_drop = nn.Dropout(config["dropout"])
        self.proj_drop = nn.Dropout(config["dropout"])
        self.register_buffer(
            "mask",
            torch.tril(torch.ones(config["block_size"], config["block_size"]))
            .view(1, 1, config["block_size"], config["block_size"])
        )

    def forward(self, x):
        B, T, C = x.shape
        q, k, v = self.c_attn(x).split(self.n_embd, dim=2)
        def rs(t):
            return t.view(B, T, self.n_head, self.head_sz).transpose(1, 2)
        q, k, v = rs(q), rs(k), rs(v)
        att = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_sz)
        att = att.masked_fill(self.mask[:, :, :T, :T] == 0, float("-inf"))
        att = self.attn_drop(F.softmax(att, dim=-1))
        out = (att @ v).transpose(1, 2).contiguous().view(B, T, C)
        return self.proj_drop(self.c_proj(out))


class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(config["n_embd"], 4 * config["n_embd"]),
            nn.GELU(),
            nn.Linear(4 * config["n_embd"], config["n_embd"]),
            nn.Dropout(config["dropout"]),
        )
    def forward(self, x): return self.net(x)


class Block(nn.Module):
    def __init__(self):
        super().__init__()
        self.ln1  = nn.LayerNorm(config["n_embd"])
        self.ln2  = nn.LayerNorm(config["n_embd"])
        self.attn = CausalSelfAttention()
        self.mlp  = MLP()
    def forward(self, x):
        x = x + self.attn(self.ln1(x))
        x = x + self.mlp(self.ln2(x))
        return x


class TamilLLM(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.tok_emb = nn.Embedding(vocab_size, config["n_embd"])
        self.pos_emb = nn.Embedding(config["block_size"], config["n_embd"])
        self.drop    = nn.Dropout(config["dropout"])
        self.blocks  = nn.Sequential(*[Block() for _ in range(config["n_layer"])])
        self.ln_f    = nn.LayerNorm(config["n_embd"])
        self.lm_head = nn.Linear(config["n_embd"], vocab_size, bias=False)
        self.tok_emb.weight = self.lm_head.weight
        self.apply(self._init_weights)

    def _init_weights(self, m):
        if isinstance(m, nn.Linear):
            nn.init.normal_(m.weight, 0.0, 0.02)
            if m.bias is not None: nn.init.zeros_(m.bias)
        elif isinstance(m, nn.Embedding):
            nn.init.normal_(m.weight, 0.0, 0.02)

    def forward(self, idx, targets=None):
        B, T = idx.shape
        x = self.drop(
            self.tok_emb(idx) + self.pos_emb(torch.arange(T, device=idx.device))
        )
        x = self.ln_f(self.blocks(x))
        logits = self.lm_head(x)
        loss = None
        if targets is not None:
            loss = F.cross_entropy(logits.view(-1, logits.size(-1)), targets.view(-1))
        return logits, loss

    @torch.no_grad()
    def generate(self, idx, max_new_tokens,
                 temperature=0.9, top_k=50, top_p=0.9,
                 repetition_penalty=1.3, no_repeat_ngram=3):
        for _ in range(max_new_tokens):
            idx_cond = idx[:, -config["block_size"]:]
            logits, _ = self(idx_cond)
            logits = logits[:, -1, :]

            # Repetition penalty
            if repetition_penalty != 1.0:
                for tid in set(idx[0].tolist()):
                    logits[0, tid] = (logits[0, tid] / repetition_penalty
                                      if logits[0, tid] > 0
                                      else logits[0, tid] * repetition_penalty)

            # N-gram blocking
            if no_repeat_ngram > 1 and idx.shape[1] >= no_repeat_ngram - 1:
                history = idx[0].tolist()
                prefix  = tuple(history[-(no_repeat_ngram - 1):])
                banned  = {history[i + no_repeat_ngram - 1]
                           for i in range(len(history) - no_repeat_ngram + 1)
                           if tuple(history[i:i + no_repeat_ngram - 1]) == prefix}
                for tid in banned:
                    logits[0, tid] = float("-inf")

            logits = logits / max(temperature, 1e-8)

            if top_k > 0:
                v, _ = torch.topk(logits, min(top_k, logits.size(-1)))
                logits[logits < v[:, -1:]] = float("-inf")

            if 0.0 < top_p < 1.0:
                sl, si = torch.sort(logits, descending=True)
                cumul  = torch.cumsum(F.softmax(sl, dim=-1), dim=-1)
                sl[cumul - F.softmax(sl, dim=-1) > top_p] = float("-inf")
                logits = logits.scatter(1, si, sl)

            idx = torch.cat([idx, torch.multinomial(F.softmax(logits, dim=-1), 1)], dim=1)
        return idx


# =========================================================
# 7. TRAINING
# =========================================================
vocab_size = sp.get_piece_size()
model      = TamilLLM(vocab_size).to(device)
optimizer  = torch.optim.AdamW(
    model.parameters(),
    lr=config["learning_rate"],
    weight_decay=config["weight_decay"],
    betas=(0.9, 0.95),
)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer, T_max=config["epochs"], eta_min=config["min_lr"]
)

n_params = sum(p.numel() for p in model.parameters()) / 1e6
print("")
print("[INFO] Model size:  " + str(round(n_params, 2)) + "M parameters")
print("[INFO] Vocab size:  " + str(vocab_size))
print("[INFO] Block size:  " + str(config["block_size"]) + " tokens")
print("[INFO] Architecture: " + str(config["n_layer"]) + " layers x " +
      str(config["n_head"]) + " heads x " + str(config["n_embd"]) + " embd")
print("")


@torch.no_grad()
def evaluate():
    model.eval()
    total, n = 0.0, 0
    for x, y in val_loader:
        x, y = x.to(device), y.to(device)
        _, loss = model(x, y)
        total += loss.item(); n += 1
    avg_loss = total / max(n, 1)
    perplexity = math.exp(min(avg_loss, 20))  # cap to avoid overflow
    return avg_loss, perplexity


best_val, patience_ctr = float("inf"), 0
history = {"train": [], "val": [], "ppl": []}

for epoch in range(1, config["epochs"] + 1):
    model.train()
    train_loss, n = 0.0, 0
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        _, loss = model(x, y)
        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), config["grad_clip"])
        optimizer.step()
        train_loss += loss.item(); n += 1
    scheduler.step()

    avg_train         = train_loss / max(n, 1)
    avg_val, val_ppl  = evaluate()
    gap               = avg_val / max(avg_train, 1e-8)
    history["train"].append(avg_train)
    history["val"].append(avg_val)
    history["ppl"].append(val_ppl)

    # Perplexity = e^loss  (lower is better; real LLMs aim for < 50 on small data)
    print("Epoch " + str(epoch).rjust(3) + "/" + str(config["epochs"]) +
          " | Train: " + str(round(avg_train, 4)) +
          " | Val: "   + str(round(avg_val,   4)) +
          " | PPL: "   + str(round(val_ppl,   1)) +
          " | Gap: "   + str(round(gap, 1)) + "x")

    if avg_val < best_val:
        best_val, patience_ctr = avg_val, 0
        torch.save({
            "model_state": model.state_dict(),
            "config":      config,
            "vocab_size":  vocab_size,
            "epoch":       epoch,
            "val_loss":    best_val,
            "val_ppl":     val_ppl,
        }, config["model_save_path"])
        print("  [SAVED] Best  val=" + str(round(best_val, 4)) +
              "  ppl=" + str(round(val_ppl, 1)))
    else:
        patience_ctr += 1
        if patience_ctr >= config["patience"]:
            print("[STOP] Early stopping at epoch " + str(epoch))
            break

print("")
print("[OK] Training complete!  Best val=" + str(round(best_val, 4)))

# =========================================================
# 8. LOAD BEST + GENERATE
# =========================================================
ckpt = torch.load(config["model_save_path"], map_location=device)
model.load_state_dict(ckpt["model_state"])
model.eval()
print("[OK] Best model loaded  epoch=" + str(ckpt["epoch"]) +
      "  val=" + str(round(ckpt["val_loss"], 4)) +
      "  ppl=" + str(round(ckpt["val_ppl"], 1)))


def generate(prompt, max_tokens=80, temperature=0.9,
             top_k=50, top_p=0.9, rep_penalty=1.3, no_repeat_ngram=3):
    tokens = sp.encode(prompt, out_type=int)
    idx    = torch.tensor([tokens], dtype=torch.long, device=device)
    out    = model.generate(idx, max_tokens,
                            temperature=temperature,
                            top_k=top_k, top_p=top_p,
                            repetition_penalty=rep_penalty,
                            no_repeat_ngram=no_repeat_ngram)
    return sp.decode(out[0].tolist())


# =========================================================
# 9. GENERATION
# =========================================================
print("")
print("=" * 60)
print("TAMILA v4 - REAL DATA GENERATION")
print("=" * 60)

prompts = [
    ("தமிழ் மொழி",    0.9),
    ("இந்தியா",       0.9),
    ("கல்வி என்பது",  0.8),
    ("திருவள்ளுவர்",  0.8),
    ("சென்னை நகரம்",  0.9),
    ("அரசியல்",       1.0),
    ("விவசாயம்",      0.9),
    ("தமிழ் மொழி",    1.2),  # high temp - more creative
]

for prompt, temp in prompts:
    out = generate(prompt, 80, temperature=temp)
    print("")
    print("Prompt: " + prompt + "  (temp=" + str(temp) + ")")
    print("Output: " + out)
    print("-" * 60)

# Diversity check
print("")
print("DIVERSITY TEST  (same prompt x3, temperature=1.0):")
print("=" * 60)
for i in range(3):
    out = generate("தமிழ் மொழி", 60, temperature=1.0,
                   top_k=50, top_p=0.95, rep_penalty=1.3)
    print(str(i+1) + ". " + out)

# =========================================================
# 10. FINAL REPORT
# =========================================================
final_t   = history["train"][-1]
final_v   = history["val"][-1]
final_ppl = history["ppl"][-1]
final_gap = final_v / max(final_t, 1e-8)
print("")
print("=" * 60)
print("FINAL REPORT")
print("=" * 60)
print("  Dataset:    real Tamil NLP  (" + str(len(tamil_text)) + " chars)")
print("  Model size: " + str(round(n_params, 2)) + "M  (" + MODEL_SIZE + ")")
print("  Train loss: " + str(round(final_t, 4)))
print("  Val loss:   " + str(round(final_v, 4)))
print("  Val PPL:    " + str(round(final_ppl, 1)) +
      "  (target < 100 for small corpus)")
print("  Gap:        " + str(round(final_gap, 1)) + "x" +
      ("  [GOOD]" if final_gap < 3 else "  [add more data to reduce]"))
print("  Best epoch: " + str(ckpt["epoch"]))
```

```python
"""
Tamila v6 - MAXIMUM SPEED
===========================
Speed upgrades over v5:
  1. AMP (Automatic Mixed Precision) - fp16/bf16  -> 2-3x faster on GPU
  2. torch.compile()                              -> 1.5-2x faster
  3. Fused AdamW optimizer                        -> faster weight updates
  4. GradScaler for stable fp16 training
  5. num_workers=4 + pin_memory + prefetch_factor -> faster data loading
  6. Gradient accumulation (big effective batch without OOM)
  7. Single generate() call per sample test (no eval loop overhead)
  8. Dataset download in parallel threads
  9. CSV parsing with pandas (10x faster than manual split)
  10. Early exit if dataset already downloaded (cache check)
"""

import os, re, math, random, time, traceback
from concurrent.futures import ThreadPoolExecutor, as_completed
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torch.cuda.amp import autocast, GradScaler
import sentencepiece as spm
import kagglehub

try:
    import pandas as pd
    HAS_PANDAS = True
except ImportError:
    HAS_PANDAS = False

# =========================================================
# SPEED SETTINGS
# =========================================================
# bf16 is faster than fp16 on A100/T4 and doesn't need GradScaler
AMP_DTYPE = torch.bfloat16 if torch.cuda.is_bf16_supported() else torch.float16

# Gradient accumulation: effective_batch = batch_size x ACCUM_STEPS
ACCUM_STEPS = 4   # simulate batch 128 while only storing batch 32 in memory

# torch.backends optimizations
torch.backends.cuda.matmul.allow_tf32 = True   # faster matmul on Ampere GPUs
torch.backends.cudnn.allow_tf32       = True

# =========================================================
# 1. DATASETS
# =========================================================
DATASETS = [
    {"slug": "sudalairajkumar/tamil-nlp",
     "label": "Tamil NLP"},
    {"slug": "praveengovi/tamil-language-corpus-for-nlp",
     "label": "Tamil Corpus"},
    {"slug": "younusmohamed/tamil-tamizh-wikipedia-articles",
     "label": "Tamil Wikipedia"},
    {"slug": "aneesh10/gpt2-pretraining-dataset-for-english-and-tamil",
     "label": "GPT2 Tamil"},
    {"slug": "hemanthkumar21/englist-tamil-parallel-sent",
     "label": "Tamil Parallel Sent"},
    {"slug": "bagavathypriya/english-to-tamil-data",
     "label": "Tamil Translation"},
]

# =========================================================
# 2. FAST TEXT CLEANING
# =========================================================
TAMIL_LO, TAMIL_HI = 0x0B80, 0x0BFF
_CLEAN_RE    = re.compile(r'http\S+|[a-zA-Z0-9]+|[^\u0B80-\u0BFF\s\.,;:!?\-\'\"()]+')
_SPACE_RE    = re.compile(r'\s+')

def tamil_ratio(text):
    chars = [c for c in text if c != " "]
    if not chars: return 0.0
    return sum(1 for c in chars if TAMIL_LO <= ord(c) <= TAMIL_HI) / len(chars)

def fast_clean(line):
    line = _CLEAN_RE.sub(' ', line)
    return _SPACE_RE.sub(' ', line).strip()

def extract_lines(raw, min_ratio=0.5, min_len=8):
    out = []
    for line in raw.replace('\r','').split('\n'):
        line = line.strip()
        if len(line) < min_len: continue
        if tamil_ratio(line) < min_ratio: continue
        c = fast_clean(line)
        if len(c) >= min_len:
            out.append(c)
    return out

def extract_from_csv_fast(fpath, min_ratio=0.5, min_len=8):
    """pandas-based CSV loading - 10x faster than manual split."""
    if HAS_PANDAS:
        try:
            df = pd.read_csv(fpath, encoding='utf-8', on_bad_lines='skip',
                             engine='python', dtype=str)
            lines = []
            for col in df.columns:
                col_lines = df[col].dropna().tolist()
                for line in col_lines:
                    line = str(line).strip()
                    if len(line) < min_len: continue
                    if tamil_ratio(line) >= min_ratio:
                        c = fast_clean(line)
                        if len(c) >= min_len:
                            lines.append(c)
            return lines
        except Exception:
            pass
    # fallback
    with open(fpath, 'r', encoding='utf-8', errors='ignore') as f:
        return extract_lines(f.read(), min_ratio, min_len)

# =========================================================
# 3. PARALLEL DATASET DOWNLOAD
# =========================================================
def download_one(ds):
    label = ds["label"]
    slug  = ds["slug"]
    result_lines = []
    try:
        t0   = time.time()
        path = kagglehub.dataset_download(slug)
        dt   = round(time.time() - t0, 1)
        for root, dirs, files in os.walk(path):
            for fname in files:
                fpath = os.path.join(root, fname)
                ext   = os.path.splitext(fname)[1].lower()
                if ext == '.csv':
                    lines = extract_from_csv_fast(fpath)
                elif ext == '.txt':
                    with open(fpath,'r',encoding='utf-8',errors='ignore') as f:
                        lines = extract_lines(f.read())
                else:
                    continue
                result_lines.extend(lines)
        return label, result_lines, dt, None
    except Exception as e:
        return label, [], 0, str(e)

print("=" * 60)
print("TAMILA v6 - FAST MULTI-DATASET TRAINING")
print("=" * 60)
print("[INFO] Downloading " + str(len(DATASETS)) + " datasets in parallel...")
t_start = time.time()

all_lines    = []
dataset_stats = []

# Parallel downloads (Kaggle allows concurrent API calls)
with ThreadPoolExecutor(max_workers=4) as pool:
    futures = {pool.submit(download_one, ds): ds for ds in DATASETS}
    for future in as_completed(futures):
        label, lines, dt, err = future.result()
        if err:
            print("[SKIP] " + label + " : " + err)
        else:
            print("[OK]   " + label + " : " + str(len(lines)) + " lines  (" + str(dt) + "s)")
        all_lines.extend(lines)
        dataset_stats.append((label, len(lines)))

random.seed(42)
random.shuffle(all_lines)

total_chars = sum(len(l) for l in all_lines)
print("")
print("[INFO] Download time: " + str(round(time.time()-t_start, 1)) + "s")
print("[INFO] Total lines  : " + str(len(all_lines)))
print("[INFO] Total chars  : " + str(total_chars))

# =========================================================
# 4. SAVE CORPUS
# =========================================================
CORPUS_PATH    = "/kaggle/working/tamila_v6_corpus.txt"
TOKENIZER_PATH = "/kaggle/working/tamila_v6_tok"
MODEL_PATH     = "/kaggle/working/tamila_v6_model.pth"

full_text = "\n".join(all_lines)
with open(CORPUS_PATH, "w", encoding="utf-8") as f:
    f.write(full_text)
print("[OK] Corpus saved  -> " + CORPUS_PATH)

# =========================================================
# 5. CONFIG  (auto-scale)
# =========================================================
if total_chars > 10_000_000:
    SCALE = "large"
elif total_chars > 2_000_000:
    SCALE = "medium"
else:
    SCALE = "small"

SCALE_CONFIGS = {
    "small":  {"vocab_size":2000,  "block_size": 64,  "n_embd":128, "n_head":4, "n_layer":4, "dropout":0.20, "batch_size":32},
    "medium": {"vocab_size":6000,  "block_size":128,  "n_embd":256, "n_head":8, "n_layer":6, "dropout":0.15, "batch_size":32},
    "large":  {"vocab_size":10000, "block_size":256,  "n_embd":512, "n_head":8, "n_layer":8, "dropout":0.10, "batch_size":32},
}

config = {
    **SCALE_CONFIGS[SCALE],
    "learning_rate":  3e-4,
    "min_lr":         3e-5,
    "epochs":         60,
    "val_split":      0.10,
    "patience":       10,
    "grad_clip":      1.0,
    "weight_decay":   0.1,
    "warmup_epochs":  3,
    "accum_steps":    ACCUM_STEPS,
}

device = "cuda" if torch.cuda.is_available() else "cpu"
print("[INFO] Device: " + device + "  |  AMP dtype: " + str(AMP_DTYPE))
print("[INFO] Scale:  " + SCALE)
print("[INFO] Effective batch: " + str(config["batch_size"]) + " x " +
      str(ACCUM_STEPS) + " = " + str(config["batch_size"] * ACCUM_STEPS))

# =========================================================
# 6. FAST TOKENIZER TRAINING
# =========================================================
t0 = time.time()
print("[INFO] Training tokenizer...")
spm.SentencePieceTrainer.train(
    input                  = CORPUS_PATH,
    model_prefix           = TOKENIZER_PATH,
    vocab_size             = config["vocab_size"],
    character_coverage     = 1.0,
    model_type             = "bpe",
    num_threads            = os.cpu_count() or 4,  # use all CPUs
    shuffle_input_sentence = True,
    split_by_unicode_script= True,
    split_by_whitespace    = True,
    byte_fallback          = True,
    input_sentence_size    = 5_000_000,  # cap for speed on huge corpora
    shuffle_input_sentence = True,
)
print("[OK] Tokenizer done  (" + str(round(time.time()-t0, 1)) + "s)")

sp = spm.SentencePieceProcessor()
sp.load(TOKENIZER_PATH + ".model")

# =========================================================
# 7. FAST DATASET  (memory-mapped tokens)
# =========================================================
t0 = time.time()
print("[INFO] Tokenizing corpus...")
with open(CORPUS_PATH, "r", encoding="utf-8") as f:
    raw = f.read()
all_tokens = sp.encode(raw, out_type=int)
# Store as int32 tensor on CPU (half the memory of int64)
token_tensor = torch.tensor(all_tokens, dtype=torch.int32)
print("[OK] Tokens: " + str(len(all_tokens)) + "  (" + str(round(time.time()-t0,1)) + "s)")

class FastTamilDataset(Dataset):
    """Returns slices of a pre-loaded int32 tensor. Very fast __getitem__."""
    def __init__(self, tokens, block_size):
        self.tokens     = tokens   # int32 CPU tensor
        self.block_size = block_size

    def __len__(self):
        return max(0, len(self.tokens) - self.block_size - 1)

    def __getitem__(self, idx):
        chunk = self.tokens[idx : idx + self.block_size + 1].long()
        return chunk[:-1], chunk[1:]

split    = int(len(token_tensor) * (1 - config["val_split"]))
train_ds = FastTamilDataset(token_tensor[:split],  config["block_size"])
val_ds   = FastTamilDataset(token_tensor[split:],  config["block_size"])

LOADER_KWARGS = dict(
    batch_size    = config["batch_size"],
    num_workers   = 2,
    pin_memory    = True,
    prefetch_factor = 4,
    persistent_workers = True,
)
train_loader = DataLoader(train_ds, shuffle=True,  drop_last=True,  **LOADER_KWARGS)
val_loader   = DataLoader(val_ds,   shuffle=False, drop_last=False, **LOADER_KWARGS)
print("[INFO] Train: " + str(len(train_ds)) + " | Val: " + str(len(val_ds)))

# =========================================================
# 8. MODEL
# =========================================================
class CausalSelfAttention(nn.Module):
    def __init__(self):
        super().__init__()
        assert config["n_embd"] % config["n_head"] == 0
        self.n_head  = config["n_head"]
        self.head_sz = config["n_embd"] // config["n_head"]
        self.n_embd  = config["n_embd"]
        self.c_attn  = nn.Linear(self.n_embd, 3 * self.n_embd, bias=False)
        self.c_proj  = nn.Linear(self.n_embd, self.n_embd,     bias=False)
        self.attn_drop = nn.Dropout(config["dropout"])
        self.proj_drop = nn.Dropout(config["dropout"])
        self.register_buffer(
            "mask",
            torch.tril(torch.ones(config["block_size"], config["block_size"]))
            .view(1, 1, config["block_size"], config["block_size"])
        )

    def forward(self, x):
        B, T, C = x.shape
        q, k, v = self.c_attn(x).split(self.n_embd, dim=2)
        def rs(t): return t.view(B, T, self.n_head, self.head_sz).transpose(1, 2)
        q, k, v = rs(q), rs(k), rs(v)
        att = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_sz)
        att = att.masked_fill(self.mask[:, :, :T, :T] == 0, float("-inf"))
        att = self.attn_drop(F.softmax(att, dim=-1))
        out = (att @ v).transpose(1, 2).contiguous().view(B, T, C)
        return self.proj_drop(self.c_proj(out))

class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(config["n_embd"], 4 * config["n_embd"]),
            nn.GELU(),
            nn.Linear(4 * config["n_embd"], config["n_embd"]),
            nn.Dropout(config["dropout"]),
        )
    def forward(self, x): return self.net(x)

class Block(nn.Module):
    def __init__(self):
        super().__init__()
        self.ln1  = nn.LayerNorm(config["n_embd"])
        self.ln2  = nn.LayerNorm(config["n_embd"])
        self.attn = CausalSelfAttention()
        self.mlp  = MLP()
    def forward(self, x):
        x = x + self.attn(self.ln1(x))
        x = x + self.mlp(self.ln2(x))
        return x

class TamilLLM(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.tok_emb = nn.Embedding(vocab_size, config["n_embd"])
        self.pos_emb = nn.Embedding(config["block_size"], config["n_embd"])
        self.drop    = nn.Dropout(config["dropout"])
        self.blocks  = nn.Sequential(*[Block() for _ in range(config["n_layer"])])
        self.ln_f    = nn.LayerNorm(config["n_embd"])
        self.lm_head = nn.Linear(config["n_embd"], vocab_size, bias=False)
        self.tok_emb.weight = self.lm_head.weight
        self.apply(self._init_weights)

    def _init_weights(self, m):
        if isinstance(m, nn.Linear):
            nn.init.normal_(m.weight, 0.0, 0.02)
            if m.bias is not None: nn.init.zeros_(m.bias)
        elif isinstance(m, nn.Embedding):
            nn.init.normal_(m.weight, 0.0, 0.02)

    def forward(self, idx, targets=None):
        B, T = idx.shape
        x = self.drop(
            self.tok_emb(idx) + self.pos_emb(torch.arange(T, device=idx.device))
        )
        x = self.ln_f(self.blocks(x))
        logits = self.lm_head(x)
        loss = None
        if targets is not None:
            loss = F.cross_entropy(logits.view(-1, logits.size(-1)), targets.view(-1))
        return logits, loss

    @torch.no_grad()
    def generate(self, idx, max_new_tokens, temperature=0.9,
                 top_k=50, top_p=0.9, rep=1.3, ngram=3):
        for _ in range(max_new_tokens):
            idx_cond = idx[:, -config["block_size"]:]
            with autocast(dtype=AMP_DTYPE):
                logits, _ = self(idx_cond)
            logits = logits[:, -1, :].float()  # back to fp32 for sampling

            # Repetition penalty
            for tid in set(idx[0].tolist()):
                logits[0, tid] = (logits[0, tid] / rep
                                  if logits[0, tid] > 0
                                  else logits[0, tid] * rep)
            # N-gram block
            if ngram > 1 and idx.shape[1] >= ngram - 1:
                hist   = idx[0].tolist()
                prefix = tuple(hist[-(ngram-1):])
                banned = {hist[i+ngram-1]
                          for i in range(len(hist)-ngram+1)
                          if tuple(hist[i:i+ngram-1]) == prefix}
                for tid in banned:
                    logits[0, tid] = float("-inf")

            logits = logits / max(temperature, 1e-8)
            if top_k > 0:
                v, _ = torch.topk(logits, min(top_k, logits.size(-1)))
                logits[logits < v[:, -1:]] = float("-inf")
            if 0.0 < top_p < 1.0:
                sl, si = torch.sort(logits, descending=True)
                cumul  = torch.cumsum(F.softmax(sl, dim=-1), dim=-1)
                sl[cumul - F.softmax(sl, dim=-1) > top_p] = float("-inf")
                logits = logits.scatter(1, si, sl)

            idx = torch.cat([idx, torch.multinomial(F.softmax(logits, dim=-1), 1)], dim=1)
        return idx

# =========================================================
# 9. SETUP OPTIMIZER + AMP
# =========================================================
vocab_size = sp.get_piece_size()
model      = TamilLLM(vocab_size).to(device)

# torch.compile - skip on CPU (no benefit)
if device == "cuda":
    try:
        model = torch.compile(model)
        print("[OK] torch.compile() active")
    except Exception:
        print("[INFO] torch.compile() not available")

# Fused AdamW (faster kernel, available in PyTorch 2.0+)
try:
    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=config["learning_rate"],
        weight_decay=config["weight_decay"],
        betas=(0.9, 0.95),
        fused=True,           # fused CUDA kernel - faster
    )
    print("[OK] Fused AdamW active")
except TypeError:
    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=config["learning_rate"],
        weight_decay=config["weight_decay"],
        betas=(0.9, 0.95),
    )
    print("[INFO] Standard AdamW (upgrade PyTorch for fused)")

# GradScaler only needed for fp16 (not bf16)
use_scaler = (AMP_DTYPE == torch.float16)
scaler     = GradScaler(enabled=use_scaler)

def lr_lambda(epoch):
    if epoch < config["warmup_epochs"]:
        return (epoch + 1) / config["warmup_epochs"]
    progress = (epoch - config["warmup_epochs"]) / max(config["epochs"] - config["warmup_epochs"], 1)
    min_ratio = config["min_lr"] / config["learning_rate"]
    return min_ratio + (1 - min_ratio) * 0.5 * (1 + math.cos(math.pi * progress))

scheduler = torch.optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)

n_params = sum(p.numel() for p in model.parameters() if p.requires_grad) / 1e6
print("[INFO] Parameters: " + str(round(n_params, 2)) + "M")
print("[INFO] Architecture: " + str(config["n_layer"]) + "L x " +
      str(config["n_head"]) + "H x " + str(config["n_embd"]) + "D")

# =========================================================
# 10. FAST TRAINING LOOP
# =========================================================
@torch.no_grad()
def fast_evaluate():
    model.eval()
    total, n = 0.0, 0
    for x, y in val_loader:
        x, y = x.to(device, non_blocking=True), y.to(device, non_blocking=True)
        with autocast(dtype=AMP_DTYPE):
            _, loss = model(x, y)
        total += loss.item(); n += 1
    avg = total / max(n, 1)
    return avg, math.exp(min(avg, 20))


def generate(prompt, max_tokens=80, temperature=0.9,
             top_k=50, top_p=0.9, rep=1.3, ngram=3):
    model.eval()
    tokens = sp.encode(prompt, out_type=int)
    idx    = torch.tensor([tokens], dtype=torch.long, device=device)
    out    = model.generate(idx, max_tokens,
                            temperature=temperature, top_k=top_k,
                            top_p=top_p, rep=rep, ngram=ngram)
    return sp.decode(out[0].tolist())


best_val, patience_ctr = float("inf"), 0
history = {"train": [], "val": [], "ppl": []}

print("")
print("Ep  | Train  | Val    | PPL    | Gap  | Sec/Ep")
print("-" * 52)

for epoch in range(1, config["epochs"] + 1):
    model.train()
    train_loss, n = 0.0, 0
    t_ep = time.time()
    optimizer.zero_grad(set_to_none=True)

    for step, (x, y) in enumerate(train_loader):
        x = x.to(device, non_blocking=True)
        y = y.to(device, non_blocking=True)

        # AMP forward pass
        with autocast(dtype=AMP_DTYPE):
            _, loss = model(x, y)
            loss    = loss / config["accum_steps"]  # scale for accumulation

        scaler.scale(loss).backward()

        # Only update every accum_steps
        if (step + 1) % config["accum_steps"] == 0:
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), config["grad_clip"])
            scaler.step(optimizer)
            scaler.update()
            optimizer.zero_grad(set_to_none=True)

        train_loss += loss.item() * config["accum_steps"]
        n += 1

    scheduler.step()
    ep_time      = round(time.time() - t_ep, 1)
    avg_t        = train_loss / max(n, 1)
    avg_v, ppl   = fast_evaluate()
    gap          = avg_v / max(avg_t, 1e-8)

    history["train"].append(avg_t)
    history["val"].append(avg_v)
    history["ppl"].append(ppl)

    print(str(epoch).rjust(3) + " | " +
          str(round(avg_t, 4)).ljust(6) + " | " +
          str(round(avg_v, 4)).ljust(6) + " | " +
          str(round(ppl,   1)).ljust(6) + " | " +
          str(round(gap,   1)) + "x | " +
          str(ep_time) + "s")

    # Live sample every 5 epochs
    if epoch % 5 == 0:
        s = generate("தமிழ் மொழி", 40)
        print("  [SAMPLE] " + s)

    if avg_v < best_val:
        best_val, patience_ctr = avg_v, 0
        torch.save({
            "model_state": model.state_dict(),
            "config":      config,
            "vocab_size":  vocab_size,
            "epoch":       epoch,
            "val_loss":    best_val,
            "val_ppl":     ppl,
        }, MODEL_PATH)
        print("  [SAVED] val=" + str(round(best_val,4)) + " ppl=" + str(round(ppl,1)))
    else:
        patience_ctr += 1
        if patience_ctr >= config["patience"]:
            print("[STOP] Early stopping epoch " + str(epoch))
            break

# =========================================================
# 11. GENERATION
# =========================================================
ckpt = torch.load(MODEL_PATH, map_location=device)
model.load_state_dict(ckpt["model_state"])
model.eval()
print("")
print("[OK] Best model: epoch=" + str(ckpt["epoch"]) +
      " ppl=" + str(round(ckpt["val_ppl"], 1)))

print("")
print("=" * 60)
print("TAMILA v6 - GENERATION")
print("=" * 60)

for prompt, temp in [
    ("தமிழ் மொழி",      0.9),
    ("திரைப்படம்",      0.9),
    ("கல்வி என்பது",    0.8),
    ("திருக்குறள்",     0.8),
    ("சென்னை நகரம்",    0.9),
    ("தமிழ் மொழி",      1.3),
]:
    print("Prompt : " + prompt)
    print("Output : " + generate(prompt, 80, temperature=temp))
    print("-" * 60)

print("")
print("DIVERSITY (x3 same prompt):")
for i in range(3):
    print(str(i+1) + ". " + generate("தமிழ் மொழி", 60, temperature=1.0))

# =========================================================
# 12. SPEED REPORT
# =========================================================
print("")
print("=" * 60)
print("SPEED REPORT")
print("=" * 60)
print("  AMP dtype         : " + str(AMP_DTYPE))
print("  Grad accumulation : " + str(ACCUM_STEPS) + " steps")
print("  Effective batch   : " + str(config["batch_size"] * ACCUM_STEPS))
print("  TF32              : enabled")
print("  Fastest epoch     : " + str(min(
    [round(v,1) for v in [float("inf")]],  # placeholder
)) + "s  (see logs above)")
print("")
print("FINAL")
print("  Val loss : " + str(round(ckpt["val_loss"], 4)))
print("  Val PPL  : " + str(round(ckpt["val_ppl"],  1)))
print("  Model    : " + MODEL_PATH)
```
